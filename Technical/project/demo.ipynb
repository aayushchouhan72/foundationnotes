Here are the most commonly used **Python string methods** with a short explanation and example.

| Method                 | Description                                 | Example                                   |
| ---------------------- | ------------------------------------------- | ----------------------------------------- |
| `upper()`              | Converts to uppercase                       | `"hello".upper()` → `"HELLO"`             |
| `lower()`              | Converts to lowercase                       | `"HELLO".lower()` → `"hello"`             |
| `capitalize()`         | First letter uppercase                      | `"python".capitalize()` → `"Python"`      |
| `title()`              | First letter of each word uppercase         | `"hello world".title()` → `"Hello World"` |
| `swapcase()`           | Swaps uppercase and lowercase               | `"PyThOn".swapcase()` → `"pYtHoN"`        |
| `casefold()`           | Stronger lowercase (for comparisons)        | `"HELLO".casefold()` → `"hello"`          |
| `strip()`              | Removes spaces from both ends               | `" hello ".strip()` → `"hello"`           |
| `lstrip()`             | Removes spaces from left                    | `" hello".lstrip()`                       |
| `rstrip()`             | Removes spaces from right                   | `"hello ".rstrip()`                       |
| `replace(old, new)`    | Replaces text                               | `"cat".replace("c","b")` → `"bat"`        |
| `find(sub)`            | Returns first index of substring            | `"python".find("t")` → `2`                |
| `rfind(sub)`           | Finds last occurrence                       | `"banana".rfind("a")` → `5`               |
| `index(sub)`           | Like `find()`, but gives error if not found | `"python".index("t")`                     |
| `count(sub)`           | Counts occurrences                          | `"banana".count("a")` → `3`               |
| `startswith(prefix)`   | Checks beginning                            | `"python".startswith("py")` → `True`      |
| `endswith(suffix)`     | Checks ending                               | `"python".endswith("on")` → `True`        |
| `split(sep)`           | Splits into a list                          | `"a,b,c".split(",")`                      |
| `rsplit(sep)`          | Splits from the right                       | `"a,b,c".rsplit(",",1)`                   |
| `splitlines()`         | Splits by new lines                         | `"a\nb".splitlines()`                     |
| `"sep".join(iterable)` | Joins list into string                      | `"-".join(["a","b"])` → `"a-b"`           |
| `partition(sep)`       | Splits into 3 parts                         | `"a-b".partition("-")`                    |
| `rpartition(sep)`      | Partition from right                        | `"a-b-c".rpartition("-")`                 |
| `center(width)`        | Centers the string                          | `"hi".center(6)`                          |
| `ljust(width)`         | Left aligns                                 | `"hi".ljust(5)`                           |
| `rjust(width)`         | Right aligns                                | `"hi".rjust(5)`                           |
| `zfill(width)`         | Pads with zeros                             | `"42".zfill(5)` → `"00042"`               |
| `isalpha()`            | Only letters?                               | `"abc".isalpha()` → `True`                |
| `isdigit()`            | Only digits?                                | `"123".isdigit()` → `True`                |
| `isnumeric()`          | Numeric characters?                         | `"123".isnumeric()`                       |
| `isalnum()`            | Letters and digits only?                    | `"abc123".isalnum()`                      |
| `isspace()`            | Only spaces?                                | `"   ".isspace()`                         |
| `islower()`            | All lowercase?                              | `"abc".islower()`                         |
| `isupper()`            | All uppercase?                              | `"ABC".isupper()`                         |
| `istitle()`            | Title case?                                 | `"Hello World".istitle()`                 |
| `isidentifier()`       | Valid Python identifier?                    | `"my_var".isidentifier()`                 |
| `isascii()`            | ASCII characters only?                      | `"abc".isascii()`                         |
| `encode()`             | Converts string to bytes                    | `"hello".encode()`                        |
| `expandtabs(tabsize)`  | Replaces tabs with spaces                   | `"a\tb".expandtabs(4)`                    |
| `format()`             | Formats a string                            | `"Hello {}".format("Ayush")`              |
| `format_map()`         | Formats using a dictionary                  | `"{name}".format_map({"name":"Ayush"})`   |
| `maketrans()`          | Creates translation table                   | `str.maketrans(...)`                      |
| `translate(table)`     | Translates characters                       | `"abc".translate(table)`                  |
| `removeprefix(prefix)` | Removes prefix                              | `"unhappy".removeprefix("un")`            |
| `removesuffix(suffix)` | Removes suffix                              | `"file.txt".removesuffix(".txt")`         |
