# terms 

### ETL

**ETL** stands for **Extract, Transform, Load**. In AI, ETL is mainly used to **prepare data before it is used for AI/ML models**.

### 1. Extract

Get data from different sources:

- MySQL / PostgreSQL
- CSV / Excel files
- APIs
- Websites
- Application databases
- Logs

Example:

```text
MySQL → customer data
CSV → sales data
API → weather data
```

### 2. Transform

Clean and prepare the extracted data.

For example:

```text
Remove duplicate records
Handle missing values
Convert data types
Normalize numerical values
Encode categorical data
Combine data from different sources
```

Example:

```text
Age: "20" → 20
Gender: "Male" → 1
Missing salary → fill with average
```

This step is especially important for **machine learning**, because models generally need structured, clean numerical/features data.

### 3. Load

Put the transformed data into a destination where it can be used.

For example:

```text
Database
Data warehouse
Data lake
Feature store
ML training dataset
```

### Simple AI example

Suppose you're building an **AI model to predict house prices**:

```text
          EXTRACT
             ↓
     MySQL + CSV + API
             ↓
         TRANSFORM
             ↓
   Clean + combine + encode
             ↓
           LOAD
             ↓
      Training dataset
             ↓
        ML Model
             ↓
     House price prediction
```

### ETL vs AI

ETL itself is **not AI**.

Think of it like:

> **ETL prepares the data; AI/ML uses that prepared data to learn and make predictions.**

For your Python/ML work, a simple ETL pipeline could look like:

```python
import pandas as pd

# Extract
df = pd.read_csv("data.csv")

# Transform
df = df.drop_duplicates()
df = df.fillna(0)

# Load
df.to_csv("clean_data.csv", index=False)
```

Here:

- `read_csv()` → **Extract**
- cleaning with pandas → **Transform**
- `to_csv()` → **Load**

# Over fitting 

### Overfitting in Machine Learning

**Overfitting** happens when a machine learning model **learns the training data too well**, including its noise and unnecessary details, instead of learning the general pattern.

As a result:

- **Training accuracy → very high**
- **Testing accuracy → low**

### Simple example

Suppose you're training a model to recognize cats and dogs.

The model sees 1,000 training images and starts memorizing specific images instead of learning general features.

```text
Training data → 99% accuracy
Testing data  → 65% accuracy
```

This is a sign of **overfitting**.

### Easy way to remember

> **Overfitting = Model memorizes instead of learning.**

### Why does overfitting happen?

Common reasons:

1. **Model is too complex**
2. **Training data is too small**
3. **Too many features**
4. **Training for too long**
5. **Data contains noise**

### How to reduce overfitting

Common techniques:

- **More training data**
- **Regularization** — L1/L2
- **Dropout** — especially in neural networks
- **Feature selection**
- **Cross-validation**
- **Early stopping**
- **Reduce model complexity**
- **Pruning** for decision trees

### Underfitting vs Overfitting

| | Underfitting | Good Fit | Overfitting |
|---|---|---|---|
| Training performance | Low | High | Very high |
| Testing performance | Low | High | Low |
| Model | Too simple | Appropriate | Too complex |
| Main problem | Doesn't learn enough | — | Memorizes data |

**Example with a decision tree:**

```text
Small tree      → Underfitting
Optimal tree    → Good fit
Very deep tree  → Overfitting
```

So for your ML notes:

> **Overfitting is a condition where a model performs very well on training data but poorly on unseen/test data because it has learned the training data too closely, including noise.**

Deep Learning is not a replacement for Machine Learning. **Deep Learning is a specialized part of Machine Learning.**

Think of it like:

```text
Artificial Intelligence (AI)
        │
        └── Machine Learning (ML)
              │
              ├── Linear Regression
              ├── Decision Tree
              ├── Random Forest
              ├── SVM
              │
              └── Deep Learning
                    ├── CNN
                    ├── RNN
                    └── Transformers
```

### So why do we need Deep Learning?

Traditional ML works very well when the data is **structured** and the important features can be selected or engineered.

For example, predicting house prices:

```text
Area
Bedrooms
Location
Age
Parking
       ↓
   ML Model
       ↓
   House Price
```

You can give these meaningful features to a model such as Random Forest or Linear Regression.

But consider **images**.

Suppose you want to detect whether an image contains a cat:

```text
Image → ??? → Cat / Not Cat
```

An image contains thousands or millions of pixels. It is difficult to manually tell traditional ML:

```text
"Look for these exact pixels to identify a cat."
```

Deep Learning can learn useful features automatically.

```text
Image
  ↓
Neural Network
  ↓
Edges
  ↓
Shapes
  ↓
Eyes / Ears / Face
  ↓
Cat
```

### Another example: voice recognition

Traditional ML:

```text
Voice
 ↓
Manually extract features
 ↓
ML algorithm
 ↓
Text
```

Deep Learning:

```text
Voice
 ↓
Deep Neural Network
 ↓
Learn useful patterns
 ↓
Text
```

### Main difference

| Machine Learning | Deep Learning |
|---|---|
| Broad field | Subfield of ML |
| Often needs feature engineering | Learns features automatically |
| Works well with structured data | Excellent for complex/unstructured data |
| Usually needs less data | Often benefits from huge datasets |
| Usually less computationally expensive | Usually requires more computation |
| Random Forest, SVM, Linear Regression | CNN, RNN, Transformers |

### Then why not use Deep Learning for everything?

Because **Deep Learning isn't always the best choice**.

For your `insurance.csv` type of dataset, where you have columns such as:

```text
age
sex
bmi
children
smoker
charges
```

A Random Forest, Decision Tree, or Linear Regression can work very well.

Using a huge neural network for this kind of relatively simple tabular data can be unnecessary.

### The simple idea

> **ML:** "Give me useful features, and I'll learn the relationship."

> **Deep Learning:** "Give me enough raw/complex data, and I'll learn useful features and the relationship."

So **Deep Learning became necessary because some problems—especially images, speech, natural language, video, and other highly complex data—were difficult to solve effectively with traditional ML approaches alone.**

### What is an Architecture Diagram?

An **architecture diagram** is a visual representation of a system that shows:

- **What components** the system has
- **How those components are connected**
- **How data flows** between them
- **Which component communicates with which**

In simple words:

> **An architecture diagram is a blueprint of a software/system that shows its major components and how they work together.**

### Example: Basic Web Application

```text
              User
               │
               ▼
        ┌─────────────┐
        │   Browser   │
        └──────┬──────┘
               │ HTTP Request
               ▼
        ┌─────────────┐
        │   Backend   │
        │  (Node.js)  │
        └──────┬──────┘
               │
       ┌───────┴────────┐
       ▼                ▼
┌─────────────┐  ┌─────────────┐
│  Database   │  │ External API│
│   MySQL     │  │             │
└─────────────┘  └─────────────┘
```

This diagram tells us that:

```text
Browser → Backend → Database
                 ↘ External API
```

### Main parts of an architecture diagram

**1. Components**

The major building blocks of the system.

Examples:

```text
Frontend
Backend
Database
API
Authentication Service
Payment Service
```

**2. Connections**

Show which components communicate with each other.

```text
Frontend ───→ Backend
Backend  ───→ Database
```

**3. Data flow**

Shows where information goes.

For example:

```text
User enters login
       ↓
Frontend
       ↓
Backend
       ↓
Database
       ↓
Backend
       ↓
Frontend
       ↓
Login successful
```

### Architecture diagram vs flowchart

They are related but different.

| Architecture Diagram | Flowchart |
|---|---|
| Shows system structure | Shows process/steps |
| Focuses on components | Focuses on decisions/actions |
| Used in system design | Used to explain algorithms/processes |
| Example: frontend → backend → database | Example: login → check password → yes/no |

### In System Design

Architecture diagrams are especially important because they help you explain **how a complete system is designed**.

For example, an e-commerce system might look like:

```text
                    Users
                      │
                      ▼
                 Load Balancer
                      │
                      ▼
                Backend Servers
             ┌────────┼────────┐
             ▼        ▼        ▼
         User API  Product API Order API
             │        │        │
             └────────┼────────┘
                      ▼
                  Database
                      │
                      ▼
                 Cache (Redis)
```

So when someone asks you **"Draw the architecture diagram of your project"**, they usually want to see the **major components + their connections + data/request flow**.

Think of it as a progression:

```text
Artificial Intelligence
        ↓
Machine Learning
        ↓
Deep Learning
        ↓
Neural Networks
        ↓
Large-scale models
        ↓
Generative AI
```

### 1. Large amount of data 📚

A Generative AI model needs huge amounts of training data.

For example, a language model can be trained on:

- Books
- Websites
- Articles
- Documents
- Code
- Conversations

The data teaches the model patterns in language and information.

### 2. Neural Network 🧠

The data is processed by a **neural network**.

Modern GenAI models commonly use architectures such as **Transformers**.

For example:

```text
Input:
"India is the"

       ↓

Neural Network

       ↓

Prediction:
"country"
```

The model learns relationships between words/tokens.

During training, the model repeatedly makes predictions and compares them with the actual data.

For example:

```text
Data
 ↓
Neural Network
 ↓
Training
 ↓
Parameters updated repeatedly
 ↓
Trained Model
```

For a language model:

```text
User: "Write a story about a robot."

             ↓

       Generative AI Model

             ↓

     Generates new text
             ↓

"A robot named Alex lived..."
```

It isn't simply copying one stored answer. It generates output based on patterns learned during training.

# Attention Is All You Need — 2017

**“Attention Is All You Need”** is a landmark research paper published in **2017** by **Ashish Vaswani and 7 other researchers**. It introduced the **Transformer architecture**, which became the foundation for many modern language models. [arXiv](https://arxiv.org/abs/1706.03762?utm_source=chatgpt.com)

[Read the original paper on arXiv](https://arxiv.org/abs/1706.03762?utm_source=chatgpt.com)

## 1. What problem did the paper solve?

Before Transformers, sequence-based tasks such as **machine translation** commonly used:

- **RNNs** — Recurrent Neural Networks
- **LSTMs** — Long Short-Term Memory
- **GRUs**
- CNN-based sequence models

The problem was that these models processed sequences largely **step by step**.

For example:

```text
"I love learning AI"

I → love → learning → AI
     ↓
   one step at a time
```

This made training difficult to parallelize and made it harder to handle relationships between words that are far apart.

The paper proposed:

> **Why not use attention itself instead of recurrence?**

That led to the **Transformer**. [NeurIPS Papers](https://papers.nips.cc/paper/2017/file/3f5ee243547dee91fbd053c1c4a845aa-Paper.pdf?utm_source=chatgpt.com)

---

# 2. The main idea: Attention

Suppose we have:

> **"The animal didn't cross the road because it was tired."**

To understand **“it”**, the model needs to determine which other words are relevant.

Attention allows the model to give different importance to different words.

Conceptually:

```text
The animal didn't cross the road because it was tired
        ↑                              ↑
        └──────── attention ───────────┘
```

The model learns **which tokens should pay attention to which other tokens**.

---

# 3. What is the Transformer?

The paper introduced this architecture:

```text
             TRANSFORMER
                  │
        ┌─────────┴─────────┐
        ↓                   ↓
     ENCODER             DECODER
        │                   │
   Self-Attention      Self-Attention
        │                   │
 Feed Forward Network  Cross Attention
        │                   │
        └─────────┬─────────┘
                  ↓
              Output
```

The original Transformer uses an **encoder-decoder architecture**. [NeurIPS Papers](https://papers.nips.cc/paper/2017/file/3f5ee243547dee91fbd053c1c4a845aa-Paper.pdf?utm_source=chatgpt.com)

---

# 4. Encoder and Decoder

### Encoder

The encoder processes the input.

For example:

```text
Input:
"How are you?"

       ↓

Encoder

       ↓

Understanding / representations
```

The encoder consists of repeated layers containing **multi-head self-attention** and **position-wise feed-forward networks**, with residual connections and layer normalization. [Hugging Face](https://huggingface.co/datasets/attention-wiki/knowledge-base/commit/1953b29bca4722a0304cdb03f684e0e39c7c5357?utm_source=chatgpt.com)

### Decoder

The decoder generates the output.

For translation:

```text
English:
"I am learning AI"

        ↓ Transformer

Hindi:
"मैं AI सीख रहा हूँ"
```

The decoder generates the output **one token at a time**, while using attention to the encoded input and previously generated tokens.

---

# 5. Self-Attention

This is the most important concept.

**Self-attention** means:

> Every word/token can look at other tokens in the same sequence to determine what information is relevant.

For example:

```text
I   love   learning   AI
│     │        │       │
└─────┴────────┴───────┘
       Attention
```

The model calculates relationships between tokens.

---

# 6. Query, Key and Value — Q, K, V

The paper's attention mechanism uses three concepts:

### Query (Q)
**What am I looking for?**

### Key (K)
**What information do I contain / what do I represent?**

### Value (V)
**What information should I provide if I am relevant?**

A useful analogy is a database search:

```text
Query
  ↓
Search Keys
  ↓
Find relevant information
  ↓
Return Values
```

The paper defines scaled dot-product attention as:

\[
Attention(Q,K,V)
=
softmax\left(\frac{QK^T}{\sqrt{d_k}}\right)V
\]

This is one of the central equations of the paper. [Hugging Face](https://huggingface.co/datasets/attention-wiki/knowledge-base/commit/1953b29bca4722a0304cdb03f684e0e39c7c5357?utm_source=chatgpt.com)

---

# 7. Why divide by √dk?

The paper calls this **scaled dot-product attention**.

Without scaling, the dot products can become large when the key dimension increases. Large values can make the softmax function produce very small gradients.

So the model divides by:

\[
\sqrt{d_k}
\]

This helps make training more stable. [Hugging Face](https://huggingface.co/datasets/attention-wiki/knowledge-base/commit/1953b29bca4722a0304cdb03f684e0e39c7c5357?utm_source=chatgpt.com)

---

# 8. Multi-Head Attention

Another major idea is **Multi-Head Attention**.

Instead of having one attention mechanism:

```text
Input
  ↓
Attention
  ↓
Output
```

the Transformer uses multiple attention heads:

```text
              Input
                ↓
      ┌─────────┼─────────┐
      ↓         ↓         ↓
    Head 1    Head 2    Head 3   ... Head N
      ↓         ↓         ↓
      └─────────┼─────────┘
                ↓
           Concatenate
                ↓
              Output
```

Different heads can learn different relationships between tokens.

For example, one head might focus more on:

```text
subject ↔ verb
```

while another can learn:

```text
pronoun ↔ noun
```

The important idea is that attention is performed in **multiple representation subspaces**.

---

# 9. Positional Encoding

There is an important problem.

Attention doesn't inherently process words sequentially like an RNN does.

Consider:

```text
"Dog bites man"
```

and

```text
"Man bites dog"
```

The words are similar, but the **order changes the meaning**.

Therefore, the Transformer adds information about the **position of each token**.

```text
Token       Position
Dog            1
bites          2
man            3
```

The original paper uses **sinusoidal positional encodings**.

So:

```text
Word Embedding
       +
Positional Encoding
       ↓
Transformer
```

---

# 10. Why was Transformer better than RNN?

The major advantage was **parallelization**.

### RNN

```text
Word 1
  ↓
Word 2
  ↓
Word 3
  ↓
Word 4
```

Processing is sequential.

### Transformer

```text
Word 1 ─┐
Word 2 ─┤
Word 3 ─┼──→ Attention
Word 4 ─┤
Word 5 ─┘
```

Tokens can be processed much more in parallel during training.

The paper specifically reports that the Transformer was more parallelizable and required significantly less training time than the recurrent approaches it compared against. [DOI](https://doi.org/10.48550/arXiv.1706.03762?utm_source=chatgpt.com)

---

# 11. The complete Transformer flow

```text
             INPUT TEXT
                 ↓
          Tokenization
                 ↓
         Word Embeddings
                 +
       Positional Encoding
                 ↓
            ENCODER
                 ↓
        Self-Attention
                 ↓
       Feed Forward Network
                 ↓
          Encoder Output
                 ↓
            DECODER
                 ↓
       Masked Self-Attention
                 ↓
       Cross-Attention
                 ↓
       Feed Forward Network
                 ↓
          Output Tokens
                 ↓
          Generated Text
```

### How Agentic AI is formed

```text
              AGENTIC AI
                  │
        ┌─────────┴─────────┐
        │                   │
   Generative AI          Tools
      (Brain)           (Hands)
        │                   │
 Understands              Executes
 reasoning               actions
        └─────────┬─────────┘
                  ↓
           Agent Controller
                  ↓
        Observe → Think → Act
                  ↑       │
                  └───────┘
```

### 1. Generative AI = Brain 🧠

The **LLM** understands the user's request and decides what needs to be done.

Example:

> "Find the weather in Indore and tell me whether I should carry an umbrella."

The GenAI model understands the task.

### 2. Tools = Ability to interact with the world 🔧

The agent can be connected to tools such as:

- Web search
- Calculator
- Database
- API
- Python/code execution
- Email
- Calendar
- File systems

For example:

```text
LLM
 ↓
Weather Tool
 ↓
Weather API
 ↓
Temperature/rain information
```

The LLM itself doesn't magically know the **current** weather. It can use a weather tool to obtain it.

### 3. Agent = Decision maker 🤖

The important part is that the system can decide:

> **Which tool should I use, and what should I do next?**

For example:

```text
User:
"Find the average salary of IT employees."

        ↓

       Agent
        ↓
"I need database information."

        ↓
    Database Tool
        ↓
      SQL Query
        ↓
    Query Result
        ↓
       Agent
        ↓
"Now calculate the average."

        ↓
      Answer
```

### 4. The Agentic AI loop

A typical agent works like this:

```text
       User Goal
           ↓
        Observe
           ↓
         Think
           ↓
    Choose an Action
           ↓
        Use Tool
           ↓
     Get Result
           ↓
        Observe
           ↓
     Think Again
           ↓
        Next Action
           ↓
        Final Answer
```

This is why it is called **agentic** — the system can pursue a goal through multiple steps rather than simply generating one response.

## MCP — Model Context Protocol

**MCP (Model Context Protocol)** is a standard that lets an **AI model connect to external tools and data sources in a structured way**.

In simple words:

> **MCP is like a standard communication bridge between an AI and external tools/data.**

### Why do we need MCP?

Suppose you have an AI agent that needs to:

- Read a file
- Query MySQL
- Search the web
- Call an API
- Access GitHub
- Send an email

Without a standard, every tool may have a different integration method.

MCP provides a common way for the AI application to discover and use those capabilities.

### Basic architecture

```text
                 AI Application
                       │
                       │ MCP
                       ↓
                ┌─────────────┐
                │ MCP Server  │
                └──────┬──────┘
                       │
          ┌────────────┼────────────┐
          ↓            ↓            ↓
       Database       Files        API
```

### Example: AI + MySQL

Imagine you ask:

> **"How many employees are in the IT department?"**

With an MCP-based setup:

```text
User
 ↓
AI Agent
 ↓
MCP Server
 ↓
MySQL Tool
 ↓
SELECT COUNT(*) ...
 ↓
Result
 ↓
AI Agent
 ↓
"3 employees are in IT."
```

The important point is that **MCP defines how the AI application communicates with the tool**.

---

## MCP has 3 important concepts

### 1. MCP Host

The AI application that wants to use MCP.

Examples could include an AI coding assistant or an AI application you build.

```text
AI Application = Host
```

### 2. MCP Client

The component inside the host that maintains communication with an MCP server.

```text
Host
 ↓
MCP Client
 ↓
MCP Server
```

### 3. MCP Server

A program that exposes capabilities to the AI application.

For example:

```text
MCP Server
 ├── Tool: query_database
 ├── Tool: search_files
 └── Tool: get_weather
```

The server can expose **tools**, **resources**, and **prompts**.

---

# MCP vs API

They are related, but not the same.

### Traditional API

```text
Your application
      ↓
    API
      ↓
Database / Service
```

You generally need to know the API's specific interface.

### MCP

```text
AI Application
      ↓
    MCP
      ↓
MCP Server
      ↓
Tool / Resource / Service
```

MCP gives AI applications a **standardized protocol for discovering and interacting with capabilities**.

# MCP + Agentic AI

This is where MCP becomes especially useful for your previous topic.

You can think of it as:

```text
             AGENTIC AI
                 │
                 ↓
             LLM / Brain
                 │
          "I need a tool"
                 │
                 ↓
                MCP
                 │
       ┌─────────┼─────────┐
       ↓         ↓         ↓
    MySQL      Files      API
     Tool       Tool      Tool
       │         │         │
       └─────────┼─────────┘
                 ↓
             Results
                 ↓
                LLM
                 ↓
              Action
```

So:

**LLM** → reasoning/generation  
**Agent** → decides what to do  
**MCP** → standardized connection layer  
**Tools** → actually perform the work

### Easy analogy

Think of an AI agent as a person:

- 🧠 **LLM** = brain
- 🔧 **Tools** = hands
- 📡 **MCP** = standard communication system between the brain and tools
- 🎯 **Agent** = decides which tool to use to accomplish the goal

**One-line exam definition:**

> **MCP (Model Context Protocol) is an open protocol that standardizes how AI applications connect to and interact with external tools, resources, and data sources.**
