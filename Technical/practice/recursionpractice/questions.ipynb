### 💡 One rule for practicing

For every recursion problem, first write these **three things** before coding:

```text
1. What is my BASE CASE?
2. What smaller problem can I solve?
3. How do I combine the current answer with the smaller answer?
```

For example, factorial:

```text
factorial(n)

Base case:
n == 0 → 1

Smaller problem:
factorial(n - 1)

Combination:
n * factorial(n - 1)
```

So:

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
```
