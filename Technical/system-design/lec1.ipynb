### Trade-off — Definition in System Design

**Trade-off:** A trade-off is a situation where **improving one aspect of a system requires sacrificing or compromising another aspect**.

**In simple words:**  
> Trade-off means choosing one benefit while giving up something else.

**Example:**  
- Increasing **security** may reduce **performance**.
- Increasing **performance** may increase **cost**.

**Short note:**  
> In system design, trade-offs are decisions made between competing requirements such as **performance, scalability, cost, security, and reliability**.
