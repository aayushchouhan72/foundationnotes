# ✍️ LanguageTool in Python

### What is LanguageTool?

**LanguageTool is a grammar, spelling, punctuation, and style-checking tool that can detect mistakes in written text.**

It can help identify:

```text
❌ Grammar mistakes
❌ Spelling mistakes
❌ Punctuation mistakes
❌ Some style problems
```

A commonly used Python package is **`language-tool-python`**.

---

# 1. Installation

Install it using:

```bash
pip install language-tool-python
```

Then import it:

```python
import language_tool_python
```

---

# 2. `language_tool_python.LanguageTool()`

### Definition:

**`LanguageTool()` creates a LanguageTool checker for a particular language.**

```python
import language_tool_python

tool = language_tool_python.LanguageTool('en-US')
```

Here:

```text
'en-US' → English (United States)
```

You can then check English text.

---

# 3. `tool.check()`

### Definition:

**`check()` checks a piece of text and returns the grammar, spelling, and style issues it finds.**

```python
import language_tool_python

tool = language_tool_python.LanguageTool('en-US')

text = "He go to school every day."

matches = tool.check(text)

print(matches)
```

LanguageTool can identify that:

```text
He go ❌
He goes ✅
```

---

# 4. Understanding `matches`

`check()` returns a list of **Match objects**.

You can loop through them:

```python
for match in matches:
    print(match)
```

A match contains information such as:

```text
Rule
Message
Position
Suggested replacements
```

---

# 5. `match.message`

### Definition:

**`message` gives a human-readable explanation of the detected mistake.**

```python
for match in matches:
    print(match.message)
```

For example, you might get a message explaining that the verb form should agree with the subject.

---

# 6. `match.replacements`

### Definition:

**`replacements` provides suggested corrections for the detected mistake.**

```python
for match in matches:
    print(match.replacements)
```

Possible result:

```text
['goes']
```

So:

```text
go → goes
```

---

# 7. `match.offset`

### Definition:

**`offset` tells you the starting character position of the detected mistake.**

```python
for match in matches:
    print(match.offset)
```

For example:

```text
3
```

means the problem starts around character index `3`.

---

# 8. `match.errorLength`

### Definition:

**`errorLength` tells you how many characters are involved in the detected error.**

```python
for match in matches:
    print(match.errorLength)
```

---

# 9. `tool.correct()`

### Definition:

**`correct()` automatically replaces detected mistakes using LanguageTool's suggestions.**

```python
import language_tool_python

tool = language_tool_python.LanguageTool('en-US')

text = "He go to school every day."

corrected = tool.correct(text)

print(corrected)
```

Possible output:

```text
He goes to school every day.
```

So:

```text
check()   → Find mistakes
correct() → Correct mistakes
```

---

# 🔥 Complete Example

Let's create a simple grammar checker:

```python
import language_tool_python

tool = language_tool_python.LanguageTool('en-US')

text = input("Enter your sentence: ")

matches = tool.check(text)

print("\nMistakes found:", len(matches))

for match in matches:
    print("\nMessage:", match.message)
    print("Suggestions:", match.replacements)

corrected = tool.correct(text)

print("\nCorrected sentence:")
print(corrected)

tool.close()
```

Example:

```text
Enter your sentence: She go to school yesterday.

Mistakes found: 1

Message: ...
Suggestions: [...]

Corrected sentence:
She went to school yesterday.
```

The exact messages and suggestions can vary by LanguageTool version/rules.

---

# 10. `tool.close()`

### Definition:

**`close()` closes the LanguageTool instance and releases its resources.**

```python
tool.close()
```

It's good practice when you're finished with the checker.
