# ⌨️🖱️ Pynput Module in Python

### What is `pynput`?

**`pynput` is a Python library used to control and monitor input devices such as the keyboard and mouse.**

It can:

* Detect keyboard presses
* Detect mouse clicks
* Detect mouse movement
* Type using Python
* Move the mouse using Python
* Click using Python

Install it:

```bash
pip install pynput
```

Then you can import:

```python
import pynput
```

But normally we use its submodules:

```python
from pynput import keyboard
from pynput import mouse
```

# ⌨️ Keyboard Module

The keyboard functionality is provided by:

```python
from pynput import keyboard
```

---

# 1. `keyboard.Listener()`

### Definition:

**`keyboard.Listener()` creates a listener that monitors keyboard events.**

```python
from pynput import keyboard

def on_press(key):
    print("Pressed:", key)

listener = keyboard.Listener(on_press=on_press)

listener.start()
```

Whenever a key is pressed, `on_press()` is called.

---

# 2. `on_press`

### Definition:

**`on_press` is a function that runs whenever a keyboard key is pressed.**

```python
def on_press(key):
    print("Key pressed:", key)
```

Example output:

```text
Key pressed: 'a'
Key pressed: 'b'
Key pressed: Key.space
```

---

# 3. `on_release`

### Definition:

**`on_release` is a function that runs whenever a keyboard key is released.**

```python
from pynput import keyboard

def on_release(key):
    print("Released:", key)

listener = keyboard.Listener(on_release=on_release)
listener.start()
```

---

# 4. Detecting a Specific Key

You can check which key was pressed.

```python
from pynput import keyboard

def on_press(key):

    if key == keyboard.Key.space:
        print("Space pressed")

    if key == keyboard.Key.enter:
        print("Enter pressed")

with keyboard.Listener(on_press=on_press) as listener:
    listener.join()
```

Some commonly used special keys:

```python
keyboard.Key.space
keyboard.Key.enter
keyboard.Key.esc
keyboard.Key.shift
keyboard.Key.ctrl
keyboard.Key.alt
keyboard.Key.tab
keyboard.Key.backspace
keyboard.Key.up
keyboard.Key.down
keyboard.Key.left
keyboard.Key.right
```

---

# 5. Normal Character Keys

For normal characters, use `key.char`.

```python
from pynput import keyboard

def on_press(key):

    try:
        print("Character:", key.char)

    except AttributeError:
        print("Special key:", key)

with keyboard.Listener(on_press=on_press) as listener:
    listener.join()
```

If you press:

```text
a
```

Output:

```text
Character: a
```

If you press:

```text
Enter
```

Output:

```text
Special key: Key.enter
```

---

# 🛑 Stopping the Listener

You can stop a listener by returning `False`.

```python
from pynput import keyboard

def on_press(key):

    print(key)

    if key == keyboard.Key.esc:
        return False

with keyboard.Listener(on_press=on_press) as listener:
    listener.join()
```

Now:

```text
Press ESC → listener stops
```

# 6. `mouse.Listener()`

### Definition:

**`mouse.Listener()` creates a listener that monitors mouse events.**

```python
from pynput import mouse

def on_click(x, y, button, pressed):

    print(x, y, button, pressed)

with mouse.Listener(on_click=on_click) as listener:
    listener.join()
```

# 7. `on_move`

### Definition:

**`on_move()` runs whenever the mouse moves.**

```python
from pynput import mouse

def on_move(x, y):
    print("Mouse:", x, y)

with mouse.Listener(on_move=on_move) as listener:
    listener.join()
```

Output might be:

```text
Mouse: 500 300
Mouse: 510 305
Mouse: 520 310
```

Here:

```text
x → horizontal position
y → vertical position
```

---

# 8. `on_click`

### Definition:

**`on_click()` runs whenever a mouse button is pressed or released.**

```python
from pynput import mouse

def on_click(x, y, button, pressed):

    if pressed:
        print("Clicked:", button)

with mouse.Listener(on_click=on_click) as listener:
    listener.join()
```

Possible output:

```text
Clicked: Button.left
```

---

# 9. `on_scroll`

### Definition:

**`on_scroll()` runs when the mouse wheel is scrolled.**

```python
from pynput import mouse

def on_scroll(x, y, dx, dy):
    print("Scroll:", dx, dy)

with mouse.Listener(on_scroll=on_scroll) as listener:
    listener.join()
```

Generally:

```text
dy > 0 → scroll up
dy < 0 → scroll down
```

# 20. Pressing a Combination

You can simulate keyboard shortcuts.

For example, `Ctrl + A`:

```python
from pynput.keyboard import Controller, Key

keyboard = Controller()

keyboard.press(Key.ctrl)
keyboard.press('a')

keyboard.release('a')
keyboard.release(Key.ctrl)
```

Or using a context manager:

```python
with keyboard.pressed(Key.ctrl):
    keyboard.press('a')
```

# 🔥 Simple Complete Example

Let's make a program that detects keyboard input and stops when you press `ESC`.

```python
from pynput import keyboard

def on_press(key):

    try:
        print("You pressed:", key.char)

    except AttributeError:
        print("Special key:", key)

    if key == keyboard.Key.esc:
        return False


with keyboard.Listener(on_press=on_press) as listener:
    listener.join()
```

### Flow:

```text
Start
  ↓
Create Listener
  ↓
Wait for keyboard input
  ↓
on_press()
  ↓
Check key
  ↓
ESC?
 ┌───────┐
Yes     No
 ↓       ↓
Stop   Continue
```

### 🎯 Important distinction

There are **two major uses of Pynput**:

```text
Pynput
  │
  ├── Listener
  │     ├── Keyboard Listener
  │     └── Mouse Listener
  │
  └── Controller
        ├── Keyboard Controller
        └── Mouse Controller
```

**Listener = monitor what the user does**

**Controller = make Python control the keyboard/mouse**

For learning, focus first on **`keyboard.Listener`**, **`mouse.Listener`**, **`keyboard.Controller`**, and **`mouse.Controller`**.
