# 👁️ What is OpenCV (`cv2`)?

**OpenCV** stands for **Open Source Computer Vision Library**. It is used to **read, modify, process, analyze, and create images and videos**, and it can also be used for tasks such as face detection and object tracking.

Install it with:

```bash
pip install opencv-python
```

Import it as:

```python
import cv2
```

> `cv2` is the name you use in Python, while the library is called **OpenCV**.

# 7. `cv2.cvtColor()`

⭐⭐⭐ Very important

**Definition:** Converts an image from one color space to another.

For example, convert BGR to grayscale:

```python
import cv2

image = cv2.imread("photo.jpg")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2.imshow("Gray Image", gray)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.COLOR_BGR2GRAY
```

→ Color image to grayscale.

```python
cv2.COLOR_BGR2RGB
```

# 8. `cv2.threshold()`

**Definition:** Converts an image into a binary/thresholded image based on a pixel-value limit.

```python
import cv2

image = cv2.imread("photo.jpg")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

ret, binary = cv2.threshold(
    gray,
    127,
    255,
    cv2.THRESH_BINARY
)

cv2.imshow("Binary", binary)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

Here:

```text
127 → threshold value
255 → maximum value
```

Pixels are separated based on whether their intensity is above or below the threshold.

# 11. `cv2.Canny()`

# 17. `camera.read()`

**Definition:** Reads the next frame from a camera or video source.

```python
success, frame = camera.read()

print(success)
```

Typically:

```text
True  → Frame successfully captured
False → Frame could not be captured
```

Complete webcam example:

```python
import cv2

camera = cv2.VideoCapture(0)

while True:

    success, frame = camera.read()

    if not success:
        break

    cv2.imshow("Camera", frame)

    if cv2.waitKey(1) == ord("q"):
        break

camera.release()
cv2.destroyAllWindows()
```

Press **Q** to exit.

# 25. `cv2.getRotationMatrix2D()`

**Definition:** Creates a transformation matrix used to rotate an image around a specified point.

```python
import cv2

image = cv2.imread("photo.jpg")

height, width = image.shape[:2]

center = (width // 2, height // 2)

matrix = cv2.getRotationMatrix2D(
    center,
    45,
    1
)
```

Here:

```text
center → rotation center
45     → angle
1      → scale
```
