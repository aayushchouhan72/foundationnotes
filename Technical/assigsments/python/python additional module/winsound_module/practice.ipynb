```python
import winsound
from pydub import AudioSegment

sound = AudioSegment.from_mp3("sec.mp3")

sound.export("sec.wav", format="wav")

winsound.PlaySound("sec.wav", winsound.SND_FILENAME)

print("Conversion completed!")
```
