# 🔊 What is `winsound`?

`winsound` is a **built-in Python module for Windows** that allows you to **play sounds and control simple system audio**.

You don't need to install it separately:

```python
import winsound
```

> ⚠️ `winsound` is **Windows-specific**. It is not available by default on Linux or macOS.

---

# 1. `winsound.Beep()`

⭐ Most important

**Definition:** Generates a simple beep sound using a specified frequency and duration.

```python
import winsound

winsound.Beep(1000, 500)
```

Here:

```text
1000 → Frequency in Hertz (Hz)
500  → Duration in milliseconds
```

So the computer produces a **1000 Hz beep for 0.5 seconds**.

### Syntax:

```python
winsound.Beep(frequency, duration)
```

Example:

```python
winsound.Beep(500, 1000)
```

→ 500 Hz sound for 1 second.

---

# 2. `winsound.PlaySound()`

⭐⭐⭐ Very important

**Definition:** Plays a sound from a WAV file or a system sound.

```python
import winsound

winsound.PlaySound("sound.wav", winsound.SND_FILENAME)
```

Here:

```text
"sound.wav"             → WAV audio file
SND_FILENAME            → tells Python that it is a filename
```

### Example with system sound

```python
winsound.PlaySound(
    "SystemAsterisk",
    winsound.SND_ALIAS
)
```

This plays a Windows system sound associated with the specified alias.

---

# 3. `winsound.MessageBeep()`

**Definition:** Plays a Windows system notification sound.

```python
import winsound

winsound.MessageBeep()
```

You can also specify a sound type:

```python
winsound.MessageBeep(winsound.MB_ICONEXCLAMATION)
```

This is useful for things such as:

```text
⚠️ Warning
❌ Error
ℹ️ Information
```

---

# 4. `winsound.SND_FILENAME`

**Definition:** A flag used with `PlaySound()` to indicate that the sound argument is a WAV filename.

```python
import winsound

winsound.PlaySound(
    "welcome.wav",
    winsound.SND_FILENAME
)
```

It tells Python:

> "The first argument is the name of a sound file."

---

# 5. `winsound.SND_ALIAS`

**Definition:** A flag used with `PlaySound()` to play a Windows system sound by its registered alias.

```python
import winsound

winsound.PlaySound(
    "SystemAsterisk",
    winsound.SND_ALIAS
)
```

---

# 6. `winsound.SND_ASYNC`

**Definition:** Plays the sound asynchronously so the program can continue running without waiting for the sound to finish.

```python
import winsound

winsound.PlaySound(
    "sound.wav",
    winsound.SND_FILENAME | winsound.SND_ASYNC
)

print("Program continues...")
```

Without `SND_ASYNC`, the program waits until the sound finishes.

Think:

```text
Normal:
Play sound → Wait → Continue

SND_ASYNC:
Play sound → Continue immediately
```

---

# 7. `winsound.SND_LOOP`

**Definition:** Repeats a sound continuously when used with asynchronous playback.

```python
import winsound

winsound.PlaySound(
    "music.wav",
    winsound.SND_FILENAME |
    winsound.SND_ASYNC |
    winsound.SND_LOOP
)
```

The sound keeps repeating.

⚠️ `SND_LOOP` should be used with `SND_ASYNC`.

---

# 8. `winsound.SND_NOSTOP`

**Definition:** Prevents a currently playing sound from being interrupted by another `PlaySound()` call.

```python
import winsound

winsound.PlaySound(
    "sound.wav",
    winsound.SND_FILENAME | winsound.SND_NOSTOP
)
```

If another sound is already playing, the new sound may not interrupt it.

---

# 9. `winsound.SND_PURGE`

**Definition:** Stops sounds that were started asynchronously by `PlaySound()`.

```python
import winsound

winsound.PlaySound(
    "music.wav",
    winsound.SND_FILENAME |
    winsound.SND_ASYNC |
    winsound.SND_LOOP
)

# Stop the sound
winsound.PlaySound(None, winsound.SND_PURGE)
```

This is particularly useful when you have a looping sound.

---

# 🎵 Complete Practical Example

Let's make a small **alarm program**:

```python
import winsound
import time

print("Alarm starting...")

for i in range(5):
    winsound.Beep(1000, 500)
    time.sleep(1)

print("Alarm finished!")
```

Here:

```text
Beep(1000, 500)
       ↓
1000 Hz frequency
500 milliseconds duration
       ↓
Wait 1 second
       ↓
Repeat 5 times
```

---

# 🔊 Playing a WAV File

```python
import winsound

winsound.PlaySound(
    "alarm.wav",
    winsound.SND_FILENAME
)
```

For background/asynchronous playback:

```python
import winsound

winsound.PlaySound(
    "alarm.wav",
    winsound.SND_FILENAME | winsound.SND_ASYNC
)

print("Sound is playing...")
```
