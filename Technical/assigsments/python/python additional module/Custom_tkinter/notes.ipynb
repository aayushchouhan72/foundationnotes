# 🖥️ CustomTkinter in Python

### What is CustomTkinter?

**CustomTkinter is a modern version of Tkinter that provides attractive and customizable GUI (Graphical User Interface) components.**

It is useful for creating:

* Login forms
* Registration forms
* Calculator applications
* Dashboards
* Desktop applications
* Buttons, labels, input fields, menus, etc.

Install it:

```bash
pip install customtkinter
```

Import it:

```python
import customtkinter as ctk
```

---

# 🔹 Basic CustomTkinter Flow

```text
Create Window
     ↓
Create Widgets
     ↓
Place Widgets
     ↓
Add Functions
     ↓
Run mainloop()
```

---

# 1. `ctk.CTk()`

### Definition:

**`CTk()` creates the main application window.**

```python
import customtkinter as ctk

app = ctk.CTk()

app.mainloop()
```

This creates a GUI window.

---

# 2. `title()`

### Definition:

**`title()` sets the title of the application window.**

```python
app.title("My Application")
```

---

# 3. `geometry()`

### Definition:

**`geometry()` sets the size of the application window.**

```python
app.geometry("500x400")
```

Meaning:

```text
500 → width
400 → height
```

---

# 4. `resizable()`

### Definition:

**`resizable()` controls whether the user can resize the window.**

```python
app.resizable(False, False)
```

Here:

```text
False → width cannot be changed
False → height cannot be changed
```

# 🔘 Buttons

# 7. `CTkButton()`

### Definition:

**`CTkButton()` creates a clickable button.**

```python
button = ctk.CTkButton(
    app,
    text="Click Me"
)

button.pack()
```

---

# 8. `command`

### Definition:

**`command` specifies the function that should execute when a button is clicked.**

```python
def hello():
    print("Button clicked")

button = ctk.CTkButton(
    app,
    text="Click Me",
    command=hello
)

button.pack()
```

⚠️ Notice:

```python
command=hello
```

not:

```python
command=hello()
```

Because we want to give the function to the button, not execute it immediately.

# 🔥 Complete Login Example

Let's combine some important widgets:

```python
import customtkinter as ctk

# Appearance
ctk.set_appearance_mode("dark")
ctk.set_default_color_theme("blue")

# Window
app = ctk.CTk()
app.title("Login Form")
app.geometry("400x300")


# Function
def login():
    username = username_entry.get()
    password = password_entry.get()

    print("Username:", username)
    print("Password:", password)


# Heading
title = ctk.CTkLabel(
    app,
    text="Login",
    font=("Arial", 25)
)

title.pack(pady=20)


# Username
username_entry = ctk.CTkEntry(
    app,
    placeholder_text="Username"
)

username_entry.pack(pady=10)


# Password
password_entry = ctk.CTkEntry(
    app,
    placeholder_text="Password",
    show="*"
)

password_entry.pack(pady=10)


# Button
login_button = ctk.CTkButton(
    app,
    text="Login",
    command=login
)

login_button.pack(pady=20)


# Start application
app.mainloop()
```

The basic GUI will look conceptually like:

```text
┌──────────────────────────┐
│          Login           │
│                          │
│   ┌──────────────────┐   │
│   │ Username         │   │
│   └──────────────────┘   │
│                          │
│   ┌──────────────────┐   │
│   │ Password         │   │
│   └──────────────────┘   │
│                          │
│       ┌──────────┐       │
│       │  Login   │       │
│       └──────────┘       │
└──────────────────────────┘
```
