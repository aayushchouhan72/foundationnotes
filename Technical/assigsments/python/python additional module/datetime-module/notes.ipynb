Yes 👍 Here is the same list with a **one-line definition + simple code example** for each important `datetime` function.

## 1. `date()`

**Definition:** Creates a date object using year, month, and day.

```python
from datetime import date

d = date(2026, 9, 12)
print(d)
```

**Output:**

```text
2026-09-12
```

---

## 2. `date.today()`

**Definition:** Returns today's current date.

```python
from datetime import date

today = date.today()
print(today)
```

**Output:**

```text
2026-09-12
```

---

## 3. `datetime()`

**Definition:** Creates an object containing both date and time.

```python
from datetime import datetime

dt = datetime(2026, 9, 12, 15, 30, 45)
print(dt)
```

**Output:**

```text
2026-09-12 15:30:45
```

---

## 4. `datetime.now()`

**Definition:** Returns the current local date and time.

```python
from datetime import datetime

now = datetime.now()
print(now)
```

**Output:**

```text
2026-09-12 20:50:25.123456
```

---

## 5. `datetime.today()`

**Definition:** Returns the current local date and time.

```python
from datetime import datetime

print(datetime.today())
```

**Output:**

```text
2026-09-12 20:50:25
```

---

## 6. `.date()`

**Definition:** Extracts only the date portion from a `datetime` object.

```python
from datetime import datetime

dt = datetime.now()

print(dt.date())
```

**Output:**

```text
2026-09-12
```

---

## 7. `.time()`

**Definition:** Extracts only the time portion from a `datetime` object.

```python
from datetime import datetime

dt = datetime.now()

print(dt.time())
```

**Output:**

```text
20:50:25.123456
```

---

## 8. `strftime()`

**Definition:** Converts a date/time object into a formatted string.

```python
from datetime import datetime

dt = datetime.now()

result = dt.strftime("%d-%m-%Y")
print(result)
```

**Output:**

```text
12-09-2026
```

Common formats:

```python
print(dt.strftime("%d/%m/%Y"))
print(dt.strftime("%B"))
print(dt.strftime("%A"))
print(dt.strftime("%I:%M:%S %p"))
```

---

## 9. `strptime()`

**Definition:** Converts a formatted string into a `datetime` object.

```python
from datetime import datetime

date_string = "12-09-2026"

dt = datetime.strptime(date_string, "%d-%m-%Y")

print(dt)
```

**Output:**

```text
2026-09-12 00:00:00
```

Remember:

```text
strftime → datetime → string
strptime → string → datetime
```

---

## 10. `timedelta()`

**Definition:** Represents a duration and is used to add or subtract time from dates.

```python
from datetime import datetime, timedelta

today = datetime.now()

future = today + timedelta(days=10)

print(future)
```

You can also use:

```python
timedelta(
    days=2,
    hours=5,
    minutes=30,
    seconds=20
)
```

---

## 11. Difference between two dates

**Definition:** Subtracting two dates calculates the time difference between them.

```python
from datetime import date

date1 = date(2026, 9, 12)
date2 = date(2026, 12, 25)

difference = date2 - date1

print(difference)
print(difference.days)
```

**Output:**

```text
104 days, 0:00:00
104
```

---

## 12. `total_seconds()`

**Definition:** Converts a `timedelta` into the total number of seconds.

```python
from datetime import timedelta

td = timedelta(days=2, hours=3)

print(td.total_seconds())
```

**Output:**

```text
183600.0
```

---

## 13. `weekday()`

**Definition:** Returns the weekday number where Monday is `0` and Sunday is `6`.

```python
from datetime import date

d = date(2026, 9, 12)

print(d.weekday())
```

**Output:**

```text
5
```

Because:

```text
Monday    = 0
Tuesday   = 1
Wednesday = 2
Thursday  = 3
Friday    = 4
Saturday  = 5
Sunday    = 6
```

---

## 14. `isoweekday()`

**Definition:** Returns the weekday number where Monday is `1` and Sunday is `7`.

```python
from datetime import date

d = date(2026, 9, 12)

print(d.isoweekday())
```

**Output:**

```text
6
```

---

## 15. `isocalendar()`

**Definition:** Returns the ISO year, week number, and weekday of a date.

```python
from datetime import date

d = date(2026, 9, 12)

info = d.isocalendar()

print(info)
```

You can get individual values:

```python
print(info.year)
print(info.week)
print(info.weekday)
```

---

## 16. `replace()`

**Definition:** Creates a new date/time object by replacing one or more components.

```python
from datetime import datetime

dt = datetime(2026, 9, 12, 15, 30)

new_dt = dt.replace(year=2030)

print(new_dt)
```

**Output:**

```text
2030-09-12 15:30:00
```

You can replace:

```python
dt.replace(year=2030)
dt.replace(month=12)
dt.replace(day=25)
dt.replace(hour=10)
```

---

## 17. `date.fromisoformat()`

**Definition:** Converts an ISO-format date string into a `date` object.

```python
from datetime import date

d = date.fromisoformat("2026-09-12")

print(d)
```

**Output:**

```text
2026-09-12
```

---

## 18. `datetime.fromisoformat()`

**Definition:** Converts an ISO-format date-time string into a `datetime` object.

```python
from datetime import datetime

dt = datetime.fromisoformat("2026-09-12 15:30:00")

print(dt)
```

**Output:**

```text
2026-09-12 15:30:00
```

---

## 19. `timestamp()`

**Definition:** Converts a `datetime` object into a Unix timestamp.

```python
from datetime import datetime

dt = datetime.now()

timestamp = dt.timestamp()

print(timestamp)
```

The timestamp represents seconds since **January 1, 1970 UTC**.

---

## 20. `fromtimestamp()`

**Definition:** Converts a Unix timestamp back into a `datetime` object.

```python
from datetime import datetime

timestamp = 1789220000

dt = datetime.fromtimestamp(timestamp)

print(dt)
```

So:

```text
timestamp → datetime
```

---

## 21. `time()`

**Definition:** Creates a time object using hour, minute, second, and microsecond.

```python
from datetime import time

t = time(15, 30, 45)

print(t)
```

**Output:**

```text
15:30:45
```

You can access:

```python
print(t.hour)
print(t.minute)
print(t.second)
```

---

## 22. `timezone`

**Definition:** Represents timezone information such as UTC.

```python
from datetime import datetime, timezone

now = datetime.now(timezone.utc)

print(now)
```
