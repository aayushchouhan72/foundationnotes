```python
import ipaddress
```

# 🌐 What is the `ipaddress` module?

**Definition:** The `ipaddress` module provides classes and functions for working with **IPv4 and IPv6 addresses, networks, and interfaces**.

For example:

```text
IPv4 → 192.168.1.10
IPv6 → 2001:db8::1
```

---

# 1. `ipaddress.ip_address()`

⭐⭐⭐ Very important

**Definition:** Creates an IPv4 or IPv6 address object from a string or integer.

```python
import ipaddress

ip = ipaddress.ip_address("192.168.1.10")

print(ip)
```

Output:

```text
192.168.1.10
```

You can check the type:

```python
print(type(ip))
```

---

## IPv6 example

```python
ip = ipaddress.ip_address("2001:db8::1")

print(ip)
```

Output:

```text
2001:db8::1
```

The function automatically determines whether the address is IPv4 or IPv6.

---

# 2. `ipaddress.ip_network()`

⭐⭐⭐ Very important

**Definition:** Creates an IP network object from a network address and prefix length.

```python
import ipaddress

network = ipaddress.ip_network("192.168.1.0/24")

print(network)
```

Output:

```text
192.168.1.0/24
```

Here:

```text
192.168.1.0 → Network address
/24          → Prefix length
```

---

# 3. `ipaddress.ip_interface()`

**Definition:** Creates an IP interface object representing an IP address together with its network prefix.

```python
import ipaddress

interface = ipaddress.ip_interface("192.168.1.10/24")

print(interface)
```

Output:

```text
192.168.1.10/24
```

Difference:

```text
ip_address()    → Address only
ip_network()    → Network
ip_interface()  → Address + network prefix
```

# 13. `.compressed`

**Definition:** Returns an IPv6 address in its shortest compressed form.

```python
import ipaddress

ip = ipaddress.ip_address(
    "2001:0db8:0000:0000:0000:0000:0000:0001"
)

print(ip.compressed)
```

```text
compressed → Short form
exploded   → Full form
```

# 25. `network.hosts()`

⭐⭐⭐ Very useful

**Definition:** Returns an iterator containing the usable host addresses in the network.

```python
import ipaddress

network = ipaddress.ip_network("192.168.1.0/30")

for ip in network.hosts():
    print(ip)
```

Output:

```text
192.168.1.1
192.168.1.2
```

For a typical IPv4 network, the network and broadcast addresses are excluded from `hosts()`.

---

# 26. `network.subnets()`

⭐⭐ Important

**Definition:** Divides a network into smaller subnetworks.

```python
import ipaddress

network = ipaddress.ip_network("192.168.1.0/24")

for subnet in network.subnets():
    print(subnet)
```

This produces smaller `/25` networks by default.

You can specify a new prefix:

```python
for subnet in network.subnets(new_prefix=26):
    print(subnet)
```

---

# 27. `network.supernet()`

**Definition:** Returns a larger network containing the current network.

```python
import ipaddress

network = ipaddress.ip_network("192.168.1.0/24")

supernet = network.supernet()

print(supernet)
```

You can specify the prefix:

```python
supernet = network.supernet(new_prefix=23)

print(supernet)
```

---

# 28. `network.overlaps()`

**Definition:** Checks whether two networks overlap.

```python
import ipaddress

network1 = ipaddress.ip_network("192.168.1.0/24")
network2 = ipaddress.ip_network("192.168.1.128/25")

print(network1.overlaps(network2))
```

Output:

```text
True
```

Because the second network lies inside the first network.

---

# 29. `network.subnet_of()`

**Definition:** Checks whether one network is completely contained inside another network.

```python
import ipaddress

network1 = ipaddress.ip_network("192.168.1.0/24")
network2 = ipaddress.ip_network("192.168.1.0/25")

print(network2.subnet_of(network1))
```

Output:

```text
True
```

---

# 30. `network.supernet_of()`

**Definition:** Checks whether one network completely contains another network.

```python
import ipaddress

network1 = ipaddress.ip_network("192.168.1.0/24")
network2 = ipaddress.ip_network("192.168.1.0/25")

print(network1.supernet_of(network2))
```

Output:

```text
True
```

# 🟢 Checking Whether an IP Belongs to a Network

This is one of the most useful operations.

**Definition:** The `in` operator checks whether an IP address belongs to a particular network.

```python
import ipaddress

ip = ipaddress.ip_address("192.168.1.50")
network = ipaddress.ip_network("192.168.1.0/24")

print(ip in network)
```

Output:

```text
True
```

Another example:

```python
ip = ipaddress.ip_address("192.168.2.50")

print(ip in network)
```

Output:

```text
False
```

### One simple example combining them

```python
import ipaddress

ip = ipaddress.ip_address("192.168.1.50")
network = ipaddress.ip_network("192.168.1.0/24")

print("IP:", ip)
print("Version:", ip.version)
print("Private:", ip.is_private)
print("Network:", network)
print("Network Address:", network.network_address)
print("Broadcast:", network.broadcast_address)
print("Total Addresses:", network.num_addresses)
print("IP belongs to network:", ip in network)
```

This is a good starting example because it shows the main purpose of the **`ipaddress` module: validating and analyzing IP addresses and networks**.
