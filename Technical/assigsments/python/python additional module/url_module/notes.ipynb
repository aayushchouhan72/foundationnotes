Absolutely. In Python, when people say **“URL module”**, they usually mean the modules used to **work with URLs**. Python's main built-in module for this is **`urllib`**.

## 🌐 Python `urllib` Module

### What is `urllib`?

**`urllib` is a built-in Python package used to work with URLs, such as creating, parsing, encoding, and making requests to web resources.**

You don't need to install it.

```python
import urllib
```

`urllib` is actually a **package** containing several useful modules:

| Module               | Purpose                                 |
| -------------------- | --------------------------------------- |
| `urllib.request`     | Open/access URLs and make HTTP requests |
| `urllib.parse`       | Parse and construct URLs                |
| `urllib.error`       | Handle URL-related errors               |
| `urllib.robotparser` | Read and check `robots.txt`             |

# 1. `urllib.parse`

This is one of the most important parts for beginners.

```python
from urllib.parse import urlparse

url = "https://www.example.com/products?id=101"

result = urlparse(url)

print(result)
```

Output:

```text
ParseResult(
 scheme='https',
 netloc='www.example.com',
 path='/products',
 params='',
 query='id=101',
 fragment=''
)
```

## 2. `urlparse()`

**Definition:** `urlparse()` breaks a URL into its different components.

```python
from urllib.parse import urlparse

url = "https://example.com/products?id=10"

x = urlparse(url)

print(x.scheme)
print(x.netloc)
print(x.path)
print(x.query)
```

Output:

```text
https
example.com
/products
id=10
```

### URL structure

```text
https://example.com/products?id=10#top
│       │           │        │
│       │           │        └── fragment
│       │           └────────── query
│       └────────────────────── path
└────────────────────────────── scheme
```

---

# 3. `urlsplit()`

**Definition:** `urlsplit()` is similar to `urlparse()` but returns the URL as five components instead of six.

```python
from urllib.parse import urlsplit

url = "https://example.com/products?id=10"

x = urlsplit(url)

print(x)
```

Output:

```text
SplitResult(
 scheme='https',
 netloc='example.com',
 path='/products',
 query='id=10',
 fragment=''
)
```

Main difference:

```text
urlparse() → 6 parts
urlsplit() → 5 parts
```

---

# 4. `urlunparse()`

**Definition:** `urlunparse()` creates a URL from its individual components.

```python
from urllib.parse import urlunparse

url = urlunparse((
    "https",
    "example.com",
    "/products",
    "",
    "id=10",
    ""
))

print(url)
```

Output:

```text
https://example.com/products?id=10
```

So:

```text
urlparse()   → URL → parts
urlunparse() → parts → URL
```

---

# 5. `urljoin()`

**Definition:** `urljoin()` combines a base URL with another URL/path.

```python
from urllib.parse import urljoin

base = "https://example.com/products/"

result = urljoin(base, "laptop")

print(result)
```

Output:

```text
https://example.com/products/laptop
```

Very useful when working with links on websites.

---

# 6. `quote()`

**Definition:** `quote()` converts special characters or spaces into URL-safe encoded form.

```python
from urllib.parse import quote

text = "hello world"

result = quote(text)

print(result)
```

Output:

```text
hello%20world
```

Why?

A URL cannot safely contain a normal space, so:

```text
space → %20
```

---

# 7. `unquote()`

**Definition:** `unquote()` converts URL-encoded characters back to normal characters.

```python
from urllib.parse import unquote

text = "hello%20world"

print(unquote(text))
```

Output:

```text
hello world
```

So:

```text
quote()   → encode
unquote() → decode
```

---

# 8. `urlencode()`

**Definition:** `urlencode()` converts a dictionary or key-value data into URL query-string format.

```python
from urllib.parse import urlencode

data = {
    "name": "Ayush",
    "age": 25
}

result = urlencode(data)

print(result)
```

Output:

```text
name=Ayush&age=25
```

This is commonly used when sending parameters to a website/API.

For example:

```text
https://example.com/search?name=Ayush&age=25
```

---

# 9. `parse_qs()`

**Definition:** `parse_qs()` converts a URL query string into a dictionary.

```python
from urllib.parse import parse_qs

query = "name=Ayush&age=25"

result = parse_qs(query)

print(result)
```

Output:

```python
{'name': ['Ayush'], 'age': ['25']}
```

Notice that the values are stored inside **lists**.

---

# 10. `parse_qsl()`

**Definition:** `parse_qsl()` converts a query string into a list of key-value tuples.

```python
from urllib.parse import parse_qsl

query = "name=Ayush&age=25"

result = parse_qsl(query)

print(result)
```

Output:

```python
[('name', 'Ayush'), ('age', '25')]
```

Difference:

```python
parse_qs()
# {'name': ['Ayush'], 'age': ['25']}

parse_qsl()
# [('name', 'Ayush'), ('age', '25')]
```

# 11. `urllib.request`

This module is used to **access resources over the internet**.

## `urlopen()`

**Definition:** `urlopen()` opens a URL and returns a response object.

```python
from urllib.request import urlopen

response = urlopen("https://example.com")

print(response.status)
```

Output:

```text
200
```

`200` generally means the request was successful.

You can read the response:

```python
from urllib.request import urlopen

response = urlopen("https://example.com")

data = response.read()

print(data)
```

The result will normally be bytes.

---

# 12. `Request()`

**Definition:** `Request()` creates a customized HTTP request.

```python
from urllib.request import Request, urlopen

req = Request(
    "https://example.com",
    headers={"User-Agent": "Mozilla/5.0"}
)

response = urlopen(req)

print(response.status)
```

This is useful when you need to specify HTTP headers.

---

# 13. `urllib.error`

This module provides exceptions for URL/network operations.

### `URLError`

**Definition:** `URLError` represents a general URL/network-related error.

```python
from urllib.request import urlopen
from urllib.error import URLError

try:
    response = urlopen("https://this-does-not-exist-example.com")
except URLError as e:
    print("Error:", e)
```

---

### `HTTPError`

**Definition:** `HTTPError` represents an HTTP error response such as `404` or `403`.

```python
from urllib.request import urlopen
from urllib.error import HTTPError

try:
    urlopen("https://example.com/not-found")
except HTTPError as e:
    print(e.code)
```

Possible output:

```text
404
```

---

# 14. `urllib.robotparser`

**Definition:** `robotparser` is used to read a website's `robots.txt` and determine whether a crawler is allowed to access a URL.

```python
from urllib.robotparser import RobotFileParser

rp = RobotFileParser("https://example.com/robots.txt")
rp.read()

print(rp.can_fetch("*", "https://example.com/"))
```

It returns:

```text
True
```

or

```text
False
```

depending on the site's `robots.txt` rules.

# ⭐ Most Important Functions to Remember

For your Python learning, focus on these first:

| Function       | One-line meaning                      |
| -------------- | ------------------------------------- |
| `urlparse()`   | Break URL into components             |
| `urlsplit()`   | Split URL into 5 components           |
| `urlunparse()` | Create URL from components            |
| `urljoin()`    | Join base URL and another path        |
| `quote()`      | Encode text for URL                   |
| `unquote()`    | Decode URL-encoded text               |
| `urlencode()`  | Convert data into query-string format |
| `parse_qs()`   | Query string → dictionary             |
| `parse_qsl()`  | Query string → list of tuples         |
| `urlopen()`    | Open/access a URL                     |
| `Request()`    | Create a customized HTTP request      |
| `URLError`     | Handle URL/network errors             |
| `HTTPError`    | Handle HTTP errors                    |

### 🔥 Easy way to remember

```text
URL
 │
 ├── parse       → URL → parts
 ├── unparse     → parts → URL
 ├── join        → combine URLs
 │
 ├── quote       → encode
 ├── unquote     → decode
 │
 ├── urlencode   → data → query string
 ├── parse_qs    → query string → dictionary
 │
 └── request
      └── urlopen() → access URL
```

If you're learning Python modules one by one, **`urllib.parse` is the best place to start**, because it teaches you how a URL is actually structured.
