# 1. Requirements Analysis

### Definition

**Requirements Analysis** software development ka ek important step hai jisme users aur stakeholders ki **needs, expectations aur requirements ko identify, understand, analyze aur document** kiya jata hai.

Simple language mein:

> **Requirements Analysis ka matlab hai ki software banane se pehle ye clearly samajhna ki user ko software se exactly kya chahiye.**

PDF mein requirements ko mainly do categories mein divide kiya gaya hai:

* Functional Requirements
* Non-Functional Requirements 

### Example

Agar hum **Online Banking System** bana rahe hain:

User ko chahiye:

* Login kar sake
* Balance check kar sake
* Money transfer kar sake
* Transaction ke baad notification mile

Ye requirements identify karna **requirements analysis** ka part hai.

### Importance

Requirements Analysis:

1. User ki needs ko clearly understand karta hai.
2. Development team ko clear direction deta hai.
3. Errors aur misunderstandings reduce karta hai.
4. Development cost reduce karne mein help karta hai.
5. Software quality improve karta hai.

### Exam Definition

> **Requirements Analysis is the process of identifying, examining, refining and documenting the needs and expectations of users and stakeholders for a software system.**

---

# 2. Functional Requirement

### Definition

**Functional Requirements** batati hain ki **system ko kya kaam karna chahiye**.

Yaad rakhne ka easiest trick:

> **Functional = WHAT system should DO**

PDF mein functional requirement ko system ki specific features aur operations ke form mein explain kiya gaya hai. 

### Example – Online Banking

Functional requirements:

* User login kar sake.
* User account balance check kar sake.
* User money transfer kar sake.
* User transaction history dekh sake.
* Transaction ke baad notification receive kare.

### Key Points

* System behavior define karta hai.
* Features/functions define karta hai.
* User requirements ko represent karta hai.
* System kya karega, ye batata hai.

### Exam Line

> **Functional requirements describe the specific functions and operations that a software system must perform to satisfy user and business needs.**

---

# 3. Non-Functional Requirement

### Definition

**Non-Functional Requirements** batati hain ki system ko **kaise perform karna chahiye**.

Easy trick:

> **Non-Functional = HOW system should perform**

PDF mein performance, security, usability, reliability aur maintainability jaise points diye gaye hain. 

### Main Types

### 1. Performance

System kitni fast response karega.

Example:

> System should respond within 2 seconds.

### 2. Security

Unauthorized users se system aur data ko protect karna.

Example:

> All banking transactions must be encrypted.

### 3. Usability

System use karne mein easy hona chahiye.

Example:

> User easily account balance check kar sake.

### 4. Reliability

System stable aur available hona chahiye.

Example:

> Banking system should be available 24×7.

### 5. Maintainability

System ko easily update aur fix kiya ja sake.

Example:

> New payment method easily add kiya ja sake.

PDF mein examples ke roop mein response time, encryption aur maximum users handle karne ki requirement bhi di gayi hai.

# 4. Functional vs Non-Functional Requirements

Ye **RGPV ke liye very important comparison question** hai.

| Functional Requirement              | Non-Functional Requirement                 |
| ----------------------------------- | ------------------------------------------ |
| Batata hai system **kya karega**    | Batata hai system **kaise perform karega** |
| Features/functions define karta hai | Quality attributes define karta hai        |
| System behavior define karta hai    | Performance/constraints define karta hai   |
| User requirements se related        | System quality requirements se related     |
| Example: Login                      | Example: Login response < 2 sec            |
| Example: Registration               | Example: Security                          |
| Example: Payment                    | Example: Reliability                       |
| Functional testing se check         | Performance/security testing se check      |

PDF ke comparison table mein bhi ye distinction diya gaya hai. 

### Shortcut

**Functional → What?**

**Non-Functional → How well?**

# 6. Requirement Elicitation

### Definition

**Requirement Elicitation** users/stakeholders se software ki requirements **identify, collect aur understand** karne ka process hai.

Ye requirement engineering ka important/initial step hai. 

### Simple Example

Suppose college Library Management System banana hai.

Developer librarian aur students se questions karega:

* Book search kaise karte ho?
* Book issue ka process kya hai?
* Fine kaise calculate hota hai?
* Overdue book ka notification chahiye?
* Admin ko book remove karne ka option chahiye?

In answers se requirements milengi.

---

# 7. Challenges in Requirement Elicitation

PDF mein major difficulties/challenges diye gaye hain. 

### 1. Unclear Requirements

Users ko khud clear nahi hota ki unhe exactly kya chahiye.

Example:

> "Mujhe ek easy application chahiye."

Easy ka meaning user aur developer ke liye different ho sakta hai.

---

### 2. Communication Gap

User aur developer ek dusre ki requirements properly understand nahi kar paate.

Example:

User technical terms nahi samajhta aur developer business process nahi samajhta.

---

### 3. Changing Requirements

Development ke during requirements change hoti rehti hain.

Example:

Initially:

> Login using password.

Later:

> Login using OTP + password.

---

### 4. Conflicting Requirements

Different stakeholders ki requirements different ho sakti hain.

Example:

* Manager wants maximum security.
* User wants very fast login.

Dono requirements ko balance karna padta hai.

# 12. Modeling

### Definition

**Modeling** software system ka ek simplified/abstract representation create karne ka process hai.

# 13. Types of Modeling

PDF mein four types diye gaye hain:

### 1. Behavioral Modeling

System external events ke response mein kaise behave karta hai.

Example:

> User login karta hai → System credentials check karta hai → Login successful/failed.

### 2. Data Flow Model

System mein data **kaise move karta hai**.

Example:

User → Login Data → Authentication → Database.

### 3. Functional Model

System ki functional requirements represent karta hai.

Example:

* Login
* Search
* Issue Book
* Return Book

### 4. Object-Oriented Model

Objects, unke attributes aur behaviors par focus karta hai.

Example:

**Book**

Attributes:

* Book ID
* Title
* Author

Behaviors:

* Issue()
* Return()

# 14. Function-Oriented Design

### Definition

Function-Oriented Design ek software design approach hai jisme system ko **small functions/modules** mein divide kiya jata hai.

Har function ek specific task perform karta hai.

### Example

Library System:

```text
Library System
       |
 ------------------
 |       |        |
Login  Search   Issue
                  |
               Return
```

PDF ke diagram mein bhi main work ko smaller functions mein divide karke dikhaya gaya hai. 

---

# 15. Characteristics of Function-Oriented Design

PDF mein ye characteristics di gayi hain. 

### 1. Functional Decomposition

Large system ko small functions/tasks mein break kiya jata hai.

### 2. Top-Down Approach

Design high-level system se start hota hai aur gradually low-level functions tak jata hai.

```text
Main System
    ↓
Major Function
    ↓
Sub Function
    ↓
Small Task
```

### 3. Sequential Flow

Functions generally specific sequence/order mein execute hote hain.

### 4. Use of DFD

**DFD = Data Flow Diagram**

DFD system ke different functions ke beech data ka movement represent karta hai.

### 5. Focus on Function

Main focus hota hai:

> **System kya function perform karega?**

rather than objects/data.

---

# 16. Advantages of Function-Oriented Design

1. Small/simple projects ke liye suitable.
2. Functions separate hone ke karan errors identify karna easier.
3. Simple aur easy to understand.



---

# 17. Disadvantages of Function-Oriented Design

PDF ke according:

1. Poor data security.
2. Complex projects ke liye suitable nahi.
3. Low code reusability.
4. Weak data handling. 

---

# 18. Object-Oriented Design

### Definition

Object-Oriented Design mein system ko **objects** ke form mein design kiya jata hai.

Har object mein generally:

* Data/Properties
* Functions/Methods

hote hain.

### Example – Car

**Object: Car**

Properties:

* Color
* Speed
* Brand

Functions:

* start()
* stop()

PDF mein exactly Car example ke through data/properties aur actions/functions explain kiye gaye hain. 

### Easy Definition

> **Object-oriented design is an approach in which software is designed using objects that combine data and functions together.**

---

# 19. Characteristics of Object-Oriented Design

### 1. Object-Based Approach

System ko real-life objects ke basis par design kiya jata hai.

Example:

Library System:

* Student
* Book
* Librarian

### 2. Abstraction

Sirf important details show karna aur unnecessary complexity hide karna.

### 3. Reusability

Classes/objects ko program ke different parts mein reuse kiya ja sakta hai.

### 4. UML Diagrams

Object-oriented design mein commonly:

* Class Diagram
* Use Case Diagram
* Sequence Diagram
* Activity Diagram

use kiye ja sakte hain.

### 5. Encapsulation

Data aur functions ko ek single unit/class mein combine karna.



---

# 20. Advantages of Object-Oriented Design

1. High code reusability.
2. Encapsulation ke through data security improve hoti hai.
3. Modular and organized design support karta hai.



---

# 21. Disadvantages of Object-Oriented Design

1. Beginners ke liye complex ho sakta hai.
2. Very small projects ke liye suitable nahi ho sakta.
3. Program size increase ho sakta hai.
4. Design aur planning mein more time lag sakta hai.

---

# 22. Function-Oriented vs Object-Oriented Design

| Function-Oriented                | Object-Oriented                         |
| -------------------------------- | --------------------------------------- |
| Functions par focus              | Objects par focus                       |
| System functions mein divide     | System objects/classes mein divide      |
| Top-down approach                | Object-based approach                   |
| Data handling comparatively weak | Data + methods together                 |
| Small projects ke liye suitable  | Large/complex projects ke liye suitable |
| Reusability relatively low       | Reusability high                        |
| DFD commonly used                | UML diagrams commonly used              |

### Memory Trick

**Function-oriented → Function first**

**Object-oriented → Object first**

# 27. Use Case Diagram Relationships

PDF ke Library Management diagram mein relationships jaise:

* Association
* `<<include>>`
* `<<extend>>`

ka use dikhaya gaya hai.

### Include

Jab ek use case ko doosre use case ki functionality **compulsorily** use karni ho.

```text
(Return Book)
      |
   <<include>>
      ↓
  (Pay Fine)
```

Example:

```text
(Update Catalogue)
       ↑
   <<extend>>
       |
   (Add Book)
```

# 33. Characteristics of Good SRS

Ye **bahut important 7-mark question** ban sakta hai.

PDF ke last page mein following characteristics di gayi hain. 

## 1. Complete

SRS mein software ki **all necessary requirements** mention honi chahiye.

> Kuch important requirement missing nahi honi chahiye.

---

## 2. Correct

Requirements customer ki actual needs ke according correct honi chahiye.

> Jo customer chahta hai wahi SRS mein specified hona chahiye.

---

## 3. Clear

Requirements clearly written honi chahiye.

Confusing language avoid karni chahiye.

---

## 4. Unambiguous

Ek requirement ka **sirf ek clear meaning** hona chahiye.

Example of ambiguous statement:

> "System should respond quickly."

"Quickly" ka exact meaning unclear hai.

Better:

> "System should respond within 2 seconds."

---

## 5. Accurate

Requirements accurate honi chahiye.

Agar SRS inaccurate hai to software correctly develop nahi ho payega.

---

## 6. Verifiable

Requirements ko testing ke through verify kiya ja sake.

Example:

> "System should respond within 2 seconds."

Tester performance test karke verify kar sakta hai.

---

## 7. Modifiable

SRS ko easily update/modify kiya ja sake jab requirements change hon.

PDF mein bhi SRS ko changes ke according easily updateable hona chahiye bataya gaya hai.

# 1. Software Design

## Definition

**Software Design** software development ka wo process hai jisme coding start hone se pehle software ke:

* Components
* Architecture
* Interfaces
* Modules
* Data structures
* Control flow

ko define kiya jata hai.

Notes mein software design ko essentially software/application banane se pehle uska **blueprint prepare karna** bataya gaya hai. 

### Simple Example

Jaise ghar banane se pehle architect:

* Rooms decide karta hai
* Doors/windows decide karta hai
* Structure banata hai

waise hi software banane se pehle **software design** kiya jata hai.

### Easy Definition for Exam

> **Software design is the process of defining the architecture, components, interfaces and other characteristics of a software system before coding.**

---

# 2. Role / Importance of Software Design in SDLC

Software Design SDLC mein **Requirements Analysis aur Coding ke beech bridge** ka kaam karta hai. 

### Importance

### 1. Reduces Errors

Coding se pehle design clear hone ke karan development errors kam hote hain.

### 2. Provides Clarity

Developers ko pata hota hai ki system kaise build karna hai.

### 3. Easy Maintenance

Properly designed software ko future mein maintain aur upgrade karna easier hota hai.

### 4. Cost and Effort Estimation

Design se development ke required effort aur cost ka better idea milta hai.

### 5. Blueprint for Developers

Design developers ko implementation ke liye roadmap provide karta hai.

### 6. Reduces Development Time and Cost

Proper planning ki wajah se unnecessary rework kam hota hai.

# 3. Software Design Process

### Definition

Simple flow:

```text
Requirements
      ↓
Software Design
      ↓
Architecture
      ↓
Modules / Components
      ↓
Detailed Design
      ↓
Coding
```

```text
Software Design Levels
        |
        |----------------------|
        ↓          ↓           ↓
   Interface   Architectural  Detailed
     Design       Design       Design
```

# 6. Detailed Design

## Definition

**Detailed Design** system ke individual components/modules ke internal details ko specify karta hai.

Yaani architecture mein hum decide karte hain:

> "Kaunse modules honge?"

Detailed design mein:

> "Har module ke andar exactly kaise kaam hoga?"

Notes ke according detailed design mein internal logic, algorithms, classes/objects aur functions/methods cover hote hain. 

### Includes

* Algorithms
* Classes
* Objects
* Functions
* Methods
* Internal logic

### Example – Payment Module

```text
Process Payment
      ↓
Verify Card
      ↓
Generate Receipt
```

---

# ⭐ Interface vs Architectural vs Detailed Design

| Interface Design             | Architectural Design               | Detailed Design                          |
| ---------------------------- | ---------------------------------- | ---------------------------------------- |
| Interaction define karta hai | Overall structure define karta hai | Internal implementation define karta hai |
| User/system interaction      | Modules/components                 | Algorithms/functions                     |
| High-level interaction       | System-level                       | Module-level                             |
| UI, screens, forms           | Components, relationships          | Classes, methods, logic                  |

### Memory Trick

**Interface → Interaction**

**Architecture → Structure**

**Detailed → Internal Logic**

# 8. Abstraction

## Definition

**Abstraction** ka matlab hai unnecessary/internal implementation details ko hide karke sirf **important/essential features** show karna.

### Simple Example

Car chalate waqt driver:

* Steering use karta hai
* Brake use karta hai
* Accelerator use karta hai

Lekin driver ko engine ke andar exactly fuel kaise burn ho raha hai, ye jaan-na necessary nahi.

Ye **abstraction** hai.

### Types in Notes

#### 1. Functional Abstraction

Function kya karta hai us par focus.

Example:

```text
sum()
```

User ko bas pata hai ki `sum()` numbers add karta hai; internal algorithm nahi pata.

#### 2. Data Abstraction

Data ki internal structure ko hide karna.

### Exam Definition

> **Abstraction is the process of hiding implementation details and showing only essential features of a system or object.**

**Modularity** ka matlab hai large software ko smaller, independent aur manageable parts called **modules** mein divide karna.

Notes mein iska purpose system ko understand, develop, test aur maintain karna easy banana diya gaya hai. 

### Example

E-commerce system:

```text
E-Commerce
   |
   |-- Login Module
   |-- Product Module
   |-- Cart Module
   |-- Payment Module
   |-- Order Module
```

### Benefits

* Easy development
* Easy testing
* Easy maintenance

# 10. Information Hiding

## Definition

**Information Hiding** ek design principle hai jisme module ki internal details ko hide kiya jata hai aur sirf necessary information expose ki jati hai.

Notes mein iska main goal data ko protect karna bataya gaya hai. 

### Example

Login system:

```text
Username → Login Module
Password → Login Module
```

User ko password database mein exactly kaise store hota hai, ye expose nahi kiya jata.

### Benefit

* Data security
* Reduced complexity
* Better maintainability

# 13. DRY Principle

### Full Form

> **DRY = Don't Repeat Yourself**

### Meaning

Same logic/code ko repeatedly write nahi karna chahiye.

Har important piece of knowledge/logic ki ideally **single representation** honi chahiye.

### Example

Agar 5 places par same tax calculation code likha hai, to future mein change karne par 5 places update karni padengi.

Better:

```text
calculateTax()
```

ek function banao aur everywhere use karo.

### Goal

> **Code duplication avoid karna.**

---

# 14. SRP

### Full Form

> **SRP = Single Responsibility Principle**

### Meaning

Ek class/module ko ideally **ek responsibility** handle karni chahiye.

### Example

Wrong:

```text
ReportPrinter
   ↓
Print Report
Save Report
Email Report
Calculate Report
```

Better:

```text
ReportPrinter → Printing
ReportSaver   → Saving
EmailService  → Email
```

Notes mein report printer class ko sirf reports print karne ki responsibility ka example diya hai. 

### Exam Line

> A class should have one reason to change.

---

# 15. ISP

### Full Form

> **ISP = Interface Segregation Principle**

### Meaning

Large interface ki jagah **small, focused interfaces** create karne chahiye.

### Example

College Management System:

```text
Student Management Interface
Faculty Management Interface
Library Management Interface
```

Har interface apne related functions handle kare.

Notes mein student, faculty aur library management groups ka example diya gaya hai. 

---

# 16. OCP

### Full Form

> **OCP = Open/Closed Principle**

### Definition

> **Software entities should be open for extension but closed for modification.**

Simple meaning:

Existing code ko repeatedly modify karne ke bajay **new functionality add/extend** karne ka design hona chahiye.

### Example

Suppose payment system mein:

```text
Payment
 ├── Credit Card
 ├── UPI
 └── Net Banking
```

Kal PayPal add karna ho to existing payment logic unnecessarily modify nahi karna chahiye; new implementation add karni chahiye.

Notes mein "new features without changing existing code" ka idea diya gaya hai. 

---

# 17. Keep It Simple

### Meaning

Software design ko unnecessarily complex nahi banana chahiye.

> **Simple design = easy understanding + easy maintenance**

### Example

Agar simple problem ko solve karne ke liye unnecessarily 10 classes bana di, to design difficult ho jayega.

Notes mein unnecessary complexity avoid karne aur design simple rakhne par focus hai. 

---

# 18. High Cohesion

### Definition

**Cohesion** measure karta hai ki ek module ke andar ke elements **kitne closely related** hain.

### High Cohesion

Agar module ke saare functions ek common purpose se related hain.

Example:

```text
Library Module
   ↓
Issue Book
Return Book
Renew Book
```

Sab library operations se related hain.

### Benefit

> **High Cohesion = Good Design**

Notes mein high cohesion ko module ke internally related elements ki degree bataya gaya hai. 

---

# 19. Coupling

### Definition

**Coupling** modules ke beech dependency ko represent karta hai.

### Low Coupling

Modules ek doosre par minimum dependent hon.

> **Low Coupling = Good Design**

### High Coupling

Modules heavily dependent hon.

> **High Coupling = Bad Design**

Example:

```text
Payment Module ──X── Notification Module
```

Payment module directly notification module ke internal details par dependent nahi hona chahiye.

Notes mein explicitly **low coupling = good design** aur high coupling = bad design diya hai. 

---

# ⭐ Cohesion vs Coupling

| Cohesion                     | Coupling                    |
| ---------------------------- | --------------------------- |
| Module ke andar relationship | Modules ke beech dependency |
| High cohesion desirable      | Low coupling desirable      |
| Related tasks together       | Modules independent         |
| Good design = High           | Good design = Low           |

### Golden Rule

> **High Cohesion + Low Coupling = Good Software Design**

# 36. Data-Centered Architecture

### Definition

Data-Centered Architecture mein ek **central data store/database** hota hai jiske around different components communicate karte hain.

Diagram:

```text
       Client
         ↕
Client ↔ Data Store ↔ Client
         ↕
       Client
```

### Main Concept

Central **Data Store**.

### Interchangeability

Notes mein component interchangeability ka point diya hai — ek component ko replace kiya ja sakta hai without unnecessarily affecting others. 

### Example

Database-centered application:

```text
Frontend
   ↕
Database
   ↕
Backend Services
```

---

# 37. Data Flow Architecture

### Definition

Data Flow Architecture mein components ko **filters** ke form mein represent kiya jata hai aur pipes ke through data transfer hota hai.

Notes mein filters aur pipes ka diagram page 18 par diya gaya hai. 

```text
Input
  ↓
Filter
  ↓
Filter
  ↓
Filter
  ↓
Output
```

### Main Components

**Filter → Data process karta hai**

**Pipe → Data transfer karta hai**

### Important Point

Data sequence of processing components se pass hota hai.

### Example

Compiler:

```text
Source Code
    ↓
Lexical Analysis
    ↓
Syntax Analysis
    ↓
Semantic Analysis
    ↓
Code Generation
```

---

# 38. Call and Return Architecture

### Definition

Call and Return Architecture mein system ko **control hierarchy** ke form mein organize kiya jata hai.

Ek main program different subprograms ko call karta hai aur subprogram execution ke baad control return karta hai.

Notes ke page 18 par iska hierarchy diagram diya hai. 

```text
             Main Program
             /     |     \
            /      |      \
      Subprogram Subprogram Subprogram
        /   \       |       /   \
      App   App    App     App  App
```

### Main Concept

> **Call → Execute → Return**

---

# 39. Object-Oriented Architecture

### Definition

Object-Oriented Architecture mein system ko **interacting objects** mein divide kiya jata hai.

Har object:

* Data
* Operations/Methods

encapsulate karta hai.

Notes ke page 19 mein classes/objects ke beech messages ke through interaction ka diagram diya gaya hai. 

Example:

```text
Class A  --message-->  Class B
   ↑                       ↓
Class C  <--message--   Class D
```

### Main Concept

Objects ek doosre ko messages bhejkar communicate karte hain.

---

# 40. Layered Architecture

### Definition

Layered Architecture mein software ko different layers mein divide kiya jata hai aur har layer specific responsibility perform karti hai.

Notes mein outer layer se UI operation aur inner/intermediate layers se services/data-related operations explain kiye gaye hain. 

Common representation:

```text
Presentation Layer
        ↓
Business Logic Layer
        ↓
Persistence Layer
        ↓
Database Layer
```

### Layers

### 1. Presentation Layer

User Interface handle karta hai.

### 2. Business Layer

Business logic handle karta hai.

### 3. Persistence Layer

Data storage/retrieval handle karta hai.

### 4. Database Layer

Actual database/data storage.

### Advantage

Har layer ki responsibility separate hone se system organized aur maintainable hota hai.

# ⭐ 49. Structured Analysis vs Structured Design

# 35. Equivalence Partitioning – EP

### Full Form

> **EP = Equivalence Partitioning**

### Definition

Input data ko different **equivalent groups/partitions** mein divide kiya jata hai, aur each group se representative test value select ki jati hai.

Notes ke page 13 par ye definition aur concept diya hai. 

### Example

```text
18 – 60
```

```text
Invalid: age < 18
Valid:   18 – 60
Invalid: age > 60
```

Representative values:

```text
10 → Invalid
30 → Valid
70 → Invalid
```

### Benefit

Kam test cases mein effective testing.

# 🧠 15 Golden Lines for Unit 4

Ye lines yaad kar lo — revision kaafi easy ho jayega:

1. **Testing = Find defects + Verify requirements.**
2. **Static Analysis = Code run kiye bina analysis.**
3. **Dynamic Analysis = Code run karke analysis.**
4. **Manual Testing = Human executes test cases.**
5. **Automation Testing = Tools/scripts execute test cases.**
6. **Testing Process = Requirement → Planning → Test Case → Environment → Execution → Defect → Closure.**
7. **Unit = Individual component.**
8. **Integration = Interaction between modules.**
9. **System = Complete system.**
10. **Acceptance = Customer/end-user acceptance.**
11. **Black Box = No internal code knowledge.**
12. **White Box = Full internal code knowledge.**
13. **Grey Box = Partial internal knowledge.**
14. **EP = Divide input into equivalent groups.**
15. **BVA = Test boundary values.**
16. **Test Oracle = Actual result vs Expected result.**
