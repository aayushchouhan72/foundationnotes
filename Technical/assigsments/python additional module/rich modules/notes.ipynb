# 🟢 1. `print()`

The most commonly used Rich function is `rich.print()`.

**Definition:** Prints text in the terminal with Rich formatting such as colors, bold, italic, and styles.

```python
from rich import print

print("[bold red]Hello World![/bold red]")
```

You can use different styles:

```python
print("[bold]Bold Text[/bold]")
print("[italic]Italic Text[/italic]")
print("[green]Success[/green]")
print("[red]Error[/red]")
print("[yellow]Warning[/yellow]")
```

---

# 🟢 2. `Console()`

**Definition:** Creates a Rich console object that provides advanced methods for displaying formatted terminal output.

```python
from rich.console import Console

console = Console()

console.print("Hello World")
```

The important point is:

```python
console = Console()
```

creates your Rich console.

---

# 🟢 3. `console.print()`

**Definition:** Prints styled and formatted content to the terminal.

```python
from rich.console import Console

console = Console()

console.print("[bold green]Success![/bold green]")
```

You can also specify styles separately:

```python
console.print("Hello", style="bold blue")
```

---

# 🟢 4. `console.log()`

**Definition:** Displays a message with useful information such as the current time and source location.

```python
from rich.console import Console

console = Console()

console.log("User logged in")
```

It produces output similar to:

```text
[14:25:10] User logged in
```

This is useful for **debugging and logging programs**.

---

# 🟢 5. `console.input()`

**Definition:** Takes user input while displaying the prompt using Rich formatting.

```python
from rich.console import Console

console = Console()

name = console.input("[bold green]Enter your name: [/bold green]")

console.print(f"Hello {name}")
```

---

# 🟢 6. `console.rule()`

**Definition:** Displays a horizontal line with an optional title.

```python
from rich.console import Console

console = Console()

console.rule("Student Information")
```

It creates a visual separator like:

```text
──────────── Student Information ────────────
```

Very useful for making terminal programs look organized.

---

# 🟢 7. `console.clear()`

**Definition:** Clears the terminal screen.

```python
from rich.console import Console

console = Console()

console.clear()
```

This is useful for programs such as:

* Menus
* Games
* Dashboards
* Repeated displays

---

# 🟢 8. `console.status()`

**Definition:** Displays a temporary status message while some operation is running.

```python
from rich.console import Console
import time

console = Console()

with console.status("Processing..."):
    time.sleep(3)

console.print("Done!")
```

While the program runs, you can see:

```text
⠋ Processing...
```

Then:

```text
Done!
```

---

# 🟢 9. `console.print_exception()`

**Definition:** Displays detailed information about the current exception in a Rich-formatted traceback.

```python
from rich.console import Console

console = Console()

try:
    x = 10 / 0
except:
    console.print_exception()
```

This is very useful for **debugging errors**.

---

# 🟢 10. `Console.size`

**Definition:** Provides the current width and height of the terminal.

```python
from rich.console import Console

console = Console()

print(console.size)
```

You can access:

```python
print(console.width)
print(console.height)
```

---

# 🟢 11. `Text`

**Definition:** Creates a Rich text object that allows different parts of a string to have different styles.

```python
from rich.text import Text

text = Text("Hello World")

text.stylize("bold red")

print(text)
```

You can style a particular section:

```python
from rich.text import Text

text = Text("Hello World")

text.stylize("bold red", 0, 5)

print(text)
```

Here, only:

```text
Hello
```

gets the style.

---

# 🟢 12. `Markdown`

**Definition:** Renders Markdown-formatted text beautifully in the terminal.

```python
from rich.console import Console
from rich.markdown import Markdown

console = Console()

md = Markdown("# Hello World\n\nThis is **bold text**.")

console.print(md)
```

Rich will render the Markdown in a formatted way.

---

# 🟢 13. `Panel`

⭐ Very useful

**Definition:** Displays content inside a visually formatted box.

```python
from rich.console import Console
from rich.panel import Panel

console = Console()

panel = Panel("Hello World")

console.print(panel)
```

You can add a title:

```python
panel = Panel(
    "Welcome to Python",
    title="Message"
)

console.print(panel)
```

---

# 🟢 14. `Table`

⭐⭐⭐ Very important

**Definition:** Creates formatted tables in the terminal.

```python
from rich.console import Console
from rich.table import Table

console = Console()

table = Table()

table.add_column("Name")
table.add_column("Age")
table.add_column("City")

table.add_row("Ayush", "25", "Bhopal")
table.add_row("Rahul", "24", "Indore")

console.print(table)
```

Output looks approximately like:

```text
┏━━━━━━━┳━━━━━┳━━━━━━━━┓
┃ Name  ┃ Age ┃ City   ┃
┡━━━━━━━╇━━━━━╇━━━━━━━━┩
│ Ayush │ 25  │ Bhopal │
│ Rahul │ 24  │ Indore │
└───────┴─────┴────────┘
```

---

# 🟢 15. `table.add_column()`

**Definition:** Adds a column to a Rich `Table`.

```python
table.add_column("Name")
table.add_column("Age")
```

Each call creates one column.

---

# 🟢 16. `table.add_row()`

**Definition:** Adds a row of data to a Rich `Table`.

```python
table.add_row("Ayush", "25")
```

If the table has:

```text
Name | Age
```

this adds:

```text
Ayush | 25
```

---

# 🟢 17. `Progress`

⭐⭐⭐ Very important

**Definition:** Creates a progress bar for displaying the progress of a long-running task.

```python
from rich.progress import track
import time

for i in track(range(10), description="Processing..."):
    time.sleep(0.5)
```

You get a progress bar such as:

```text
Processing... ━━━━━━━━━━━━━━━━ 100%
```

---

# 🟢 18. `track()`

**Definition:** Wraps an iterable and automatically displays its progress while it is being processed.

```python
from rich.progress import track
import time

for item in track(range(20)):
    time.sleep(0.1)
```

This is one of the easiest ways to add a progress bar to a loop.

---

# 🟢 19. `Spinner`

**Definition:** Displays an animated spinner while an operation is running.

Example:

```python
from rich.console import Console
from rich.spinner import Spinner

console = Console()

spinner = Spinner("dots", text="Loading...")

console.print(spinner)
```

---

# 🟢 20. `Live`

**Definition:** Allows terminal output to be updated continuously without repeatedly printing new lines.

Example:

```python
from rich.live import Live
from rich.console import Group
import time

with Live("Loading...", refresh_per_second=4):
    time.sleep(5)
```

This is useful for:

* Live dashboards
* Clocks
* Counters
* System monitoring

---

# 🟢 21. `Pretty`

**Definition:** Displays Python objects such as dictionaries and lists in a clean, readable format.

```python
from rich import print
from rich.pretty import Pretty

data = {
    "name": "Ayush",
    "age": 25,
    "skills": ["Python", "SQL", "AI"]
}

print(Pretty(data))
```

This is especially useful when displaying **complex data structures**.

---

# 🟢 22. `inspect()`

**Definition:** Displays detailed information about a Python object, including its attributes and methods.

```python
from rich import inspect

data = [1, 2, 3]

inspect(data)
```

It can help you understand:

```text
What is this object?
What methods does it have?
What attributes does it have?
```

Very useful while **learning and debugging Python**.

---

# 🟢 23. `Traceback`

**Definition:** Provides Rich-formatted error tracebacks that are easier to read and understand.

```python
from rich.traceback import install

install()

x = 10 / 0
```

Instead of the normal traceback, Rich provides a much more readable traceback.

---

# 🟢 24. `Syntax`

**Definition:** Displays source code in the terminal with syntax highlighting.

```python
from rich.console import Console
from rich.syntax import Syntax

console = Console()

code = '''
def hello():
    print("Hello World")
'''

syntax = Syntax(code, "python")

console.print(syntax)
```

Python keywords, strings, functions, etc. are visually highlighted.
