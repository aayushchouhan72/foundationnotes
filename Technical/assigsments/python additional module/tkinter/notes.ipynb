# 🖥️ What is Tkinter?

**Tkinter** is Python's standard library for creating **Graphical User Interfaces (GUI)**.

Without Tkinter, a normal Python program might look like:

```text
Enter your name: Ayush
Enter your age: 25
```

With Tkinter, you can create an actual window:

```text
┌─────────────────────────────┐
│       Student Form          │
│                             │
│ Name:  [____________]       │
│ Age:   [____________]       │
│                             │
│        [ Submit ]            │
└─────────────────────────────┘
```

So, **Tkinter is used when you want your Python program to have windows, buttons, text boxes, menus, forms, etc.**

# 🔥 Basic Tkinter Structure

The basic structure is:

```python
import tkinter as tk

root = tk.Tk()

root.mainloop()
```

There are three important steps:

```text
import tkinter
      ↓
create window
      ↓
mainloop()
```

---

# 1. `Tk()`

⭐ Most important

**Definition:** Creates the main/root window of a Tkinter application.

```python
import tkinter as tk

root = tk.Tk()

root.mainloop()
```

You can give the window a title:

```python
root.title("My First Application")
```

---

# 2. `mainloop()`

⭐ Very important

**Definition:** Starts Tkinter's event loop and keeps the GUI window running and responsive.

```python
import tkinter as tk

root = tk.Tk()

root.title("My App")

root.mainloop()
```

Without:

```python
root.mainloop()
```

the window will usually appear and immediately close.

Think of `mainloop()` as:

```text
Keep the application running
       ↓
Wait for user actions
       ↓
Button clicked?
       ↓
Mouse moved?
       ↓
Keyboard pressed?
       ↓
Handle the event
```

---

# 3. `title()`

**Definition:** Sets the title displayed in the window's title bar.

```python
import tkinter as tk

root = tk.Tk()

root.title("Student Management System")

root.mainloop()
```

---

# 4. `geometry()`

⭐ Very useful

**Definition:** Sets the size and optionally the position of the Tkinter window.

```python
import tkinter as tk

root = tk.Tk()

root.geometry("500x400")

root.mainloop()
```

Here:

```text
500 → width
400 → height
```

You can also specify position:

```python
root.geometry("500x400+100+50")
```

Meaning approximately:

```text
500  = width
400  = height
100  = distance from left
50   = distance from top
```

---

# 5. `resizable()`

**Definition:** Controls whether the user can resize the window horizontally and vertically.

```python
import tkinter as tk

root = tk.Tk()

root.geometry("500x400")

root.resizable(False, False)

root.mainloop()
```

Here:

```text
False → Cannot resize horizontally
False → Cannot resize vertically
```

You can allow only horizontal resizing:

```python
root.resizable(True, False)
```

---

# 6. `minsize()`

**Definition:** Sets the minimum width and height allowed for the window.

```python
root.minsize(300, 200)
```

The user cannot make the window smaller than:

```text
300 × 200
```

---

# 7. `maxsize()`

**Definition:** Sets the maximum width and height allowed for the window.

```python
root.maxsize(1000, 700)
```

---

# 8. `configure()` / `config()`

**Definition:** Changes the properties or configuration of a Tkinter widget.

Example:

```python
import tkinter as tk

root = tk.Tk()

root.configure(bg="lightblue")

root.mainloop()
```

`config()` is a shorter form:

```python
root.config(bg="lightblue")
```

# 🟢 Tkinter Widgets

A **widget** is a GUI component such as:

```text
Button
Label
Entry
Text box
Checkbutton
Radio button
Listbox
Menu
Frame
```

---

# 9. `Label()`

⭐ Very important

**Definition:** Creates a widget for displaying text or images.

```python
import tkinter as tk

root = tk.Tk()

label = tk.Label(root, text="Hello World")

label.pack()

root.mainloop()
```

You can change its appearance:

```python
label = tk.Label(
    root,
    text="Welcome",
    font=("Arial", 20),
    fg="blue",
    bg="yellow"
)

label.pack()
```

Important options:

```text
text → Text to display
font → Font style/size
fg   → Foreground/text color
bg   → Background color
```

---

# 10. `Button()`

⭐ Very important

**Definition:** Creates a clickable button that can execute a function.

```python
import tkinter as tk

def hello():
    print("Button clicked!")

root = tk.Tk()

button = tk.Button(
    root,
    text="Click Me",
    command=hello
)

button.pack()

root.mainloop()
```

When the user clicks the button:

```text
Button clicked!
```

### Important:

```python
command=hello
```

NOT:

```python
command=hello()
```

Because `command` expects the function itself.

---

# 11. `Entry()`

⭐⭐⭐ Very important

**Definition:** Creates a single-line text box for accepting user input.

```python
import tkinter as tk

root = tk.Tk()

entry = tk.Entry(root)

entry.pack()

root.mainloop()
```

The user can type:

```text
Ayush
```

### Getting the entered value

```python
name = entry.get()

print(name)
```

---

# 12. `insert()`

**Definition:** Inserts text into an `Entry` widget at a specified position.

```python
entry.insert(0, "Ayush")
```

Here:

```text
0 → Start of the Entry
```

---

# 13. `delete()`

**Definition:** Removes characters from an `Entry` widget.

```python
entry.delete(0, tk.END)
```

This removes everything from the Entry.

---

# 14. `Text()`

**Definition:** Creates a multi-line text input/display widget.

```python
import tkinter as tk

root = tk.Tk()

text = tk.Text(root, height=10, width=40)

text.pack()

root.mainloop()
```

Unlike `Entry`, `Text` supports multiple lines.

```text
Entry → one line
Text  → multiple lines
```

---

# 15. `Checkbutton()`

**Definition:** Creates a checkbox that allows the user to select or deselect an option.

```python
import tkinter as tk

root = tk.Tk()

check = tk.Checkbutton(
    root,
    text="I agree"
)

check.pack()

root.mainloop()
```

It looks approximately like:

```text
☐ I agree
```

---

# 16. `IntVar()`

**Definition:** Creates a Tkinter variable object that stores an integer value and can be connected to a widget.

```python
import tkinter as tk

root = tk.Tk()

value = tk.IntVar()

check = tk.Checkbutton(
    root,
    text="I agree",
    variable=value
)

check.pack()

root.mainloop()
```

You can get its value:

```python
print(value.get())
```

Typically:

```text
0 → unchecked
1 → checked
```

---

# 17. `StringVar()`

⭐ Very useful

**Definition:** Creates a Tkinter variable that stores a string and can be connected to widgets.

```python
import tkinter as tk

root = tk.Tk()

name = tk.StringVar()

entry = tk.Entry(root, textvariable=name)

entry.pack()

root.mainloop()
```

Get value:

```python
print(name.get())
```

Set value:

```python
name.set("Ayush")
```

---

# 18. `DoubleVar()`

**Definition:** Creates a Tkinter variable for storing floating-point values.

```python
import tkinter as tk

root = tk.Tk()

price = tk.DoubleVar()

price.set(99.50)

print(price.get())

root.mainloop()
```

---

# 19. `BooleanVar()`

**Definition:** Creates a Tkinter variable for storing `True` or `False`.

```python
import tkinter as tk

root = tk.Tk()

status = tk.BooleanVar()

status.set(True)

print(status.get())

root.mainloop()
```

---

# 20. `Radiobutton()`

⭐ Important

**Definition:** Creates a button that allows the user to select one option from a group.

```python
import tkinter as tk

root = tk.Tk()

gender = tk.StringVar()

tk.Radiobutton(
    root,
    text="Male",
    variable=gender,
    value="Male"
).pack()

tk.Radiobutton(
    root,
    text="Female",
    variable=gender,
    value="Female"
).pack()

root.mainloop()
```

The idea is:

```text
○ Male
○ Female
```

Usually only one can be selected when they share the same `variable`.

# 24. `listbox.delete()`

# 25. `Frame()`

⭐ Important

**Definition:** Creates a container used to group and organize other widgets.

```python
import tkinter as tk

root = tk.Tk()

frame = tk.Frame(root, bg="lightblue")

frame.pack()

label = tk.Label(frame, text="Hello")

label.pack()

root.mainloop()
```

# 🟢 Geometry Managers

Tkinter provides three main ways to position widgets.

```text
pack()
grid()
place()
```

---

# 26. `pack()`

**Definition:** Automatically arranges widgets along the top, bottom, left, or right side of their parent.

```python
label.pack()
```

Example:

```python
label.pack(pady=20)
```

`pady` adds vertical spacing.

---

# 27. `grid()`

⭐⭐⭐ Very important

**Definition:** Arranges widgets in rows and columns like a table.

```python
import tkinter as tk

root = tk.Tk()

tk.Label(root, text="Name").grid(row=0, column=0)
tk.Entry(root).grid(row=0, column=1)

tk.Label(root, text="Age").grid(row=1, column=0)
tk.Entry(root).grid(row=1, column=1)

root.mainloop()
```

It creates something like:

```text
Name : [__________]

Age  : [__________]
```

---

# 28. `place()`

**Definition:** Positions widgets at specific x and y coordinates.

```python
label = tk.Label(root, text="Hello")

label.place(x=100, y=50)
```

Here:

```text
x = horizontal position
y = vertical position
```

---

# ⚠️ Important Rule

Generally, **don't use `pack()` and `grid()` in the same parent container**.

For example, avoid:

```python
label.pack()
button.grid(row=1, column=1)
```

in the same parent.

You can, however, use them in **different Frames**.

# 33. `Menu()`

**Definition:** Creates a menu bar or dropdown menu for a Tkinter application.

```python
import tkinter as tk

root = tk.Tk()

menu = tk.Menu(root)

root.config(menu=menu)

file_menu = tk.Menu(menu)

menu.add_cascade(
    label="File",
    menu=file_menu
)

file_menu.add_command(label="New")
file_menu.add_command(label="Open")
file_menu.add_command(label="Exit")

root.mainloop()
```

You get something like:

```text
File
 ├── New
 ├── Open
 └── Exit
```

# 🟢 Event Handling

Tkinter is **event-driven**.

That means the program waits for events:

```text
Mouse click
Keyboard press
Button click
Window close
Mouse movement
```

---

# 34. `bind()`

**Definition:** Connects a specific event, such as a key press or mouse click, to a function.

```python
import tkinter as tk

root = tk.Tk()

def clicked(event):
    print("Mouse clicked")

root.bind("<Button-1>", clicked)

root.mainloop()
```

`<Button-1>` means the **left mouse button**.

---

# 35. `after()`

⭐ Very useful

**Definition:** Schedules a function to run after a specified number of milliseconds.

```python
import tkinter as tk

root = tk.Tk()

def hello():
    print("Hello")

root.after(3000, hello)

root.mainloop()
```

After:

```text
3000 milliseconds = 3 seconds
```

the function executes.

This is useful for:

* Digital clocks
* Timers
* Animations
* Automatic updates

---

# 🟢 Window Closing

# 36. `destroy()`

**Definition:** Closes/destroys a Tkinter window or widget.

```python
import tkinter as tk

root = tk.Tk()

button = tk.Button(
    root,
    text="Close",
    command=root.destroy
)

button.pack()

root.mainloop()
```

When the button is clicked, the window closes.
