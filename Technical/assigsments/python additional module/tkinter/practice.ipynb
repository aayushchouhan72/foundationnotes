```python
import tkinter as tk
def hello():
    print("Button Clicked ")

root =  tk.Tk()


root.title("Aayush ")
root.geometry("600x600+300+90")
label = tk.Label(
    root,
    text="Wellcome..",
    font=("Arial",20),
    fg="blue",
    bg="yellow"
)
label.pack()
button=tk.Button(
    root,
    text="Click me",
    command=hello
)
button.pack()

root.mainloop()
```

```python
import tkinter as tk 
def hello(hii):
    print("Button Clicked ",hii)
root=tk.Tk()
root.title("Aayush ")
root.geometry("600x600+300+90")
label = tk.Label(
    root,
    text="Wellcome..",
    font=("Arial",20),
    fg="blue",
    bg="yellow"
)
label.pack()

entry =  tk.Entry(root,text="enter the text ..")

entry.pack()

button=tk.Button(
    root,
    text="Click me",
    command=hello(entry.get())
)
button.pack()
root.mainloop()
```
