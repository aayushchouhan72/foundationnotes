# 🗄️ SQLite in Python

### What is SQLite?

**SQLite is a lightweight, serverless database that stores data in a single file.**

Python already provides the built-in **`sqlite3` module**, so you don't need to install anything.

```python
import sqlite3
```

SQLite is useful for applications where you need to store structured data such as:

```text
Student records
Patient records
Employee records
Products
Login information
Orders
```

---

# 🔹 Basic SQLite Flow

Almost every SQLite program follows this pattern:

```text
Connect
   ↓
Create Cursor
   ↓
Execute SQL
   ↓
Commit
   ↓
Fetch Data
   ↓
Close Connection
```

---

# 1. `sqlite3.connect()`

### Definition:

**`connect()` creates a connection to an SQLite database.**

```python
import sqlite3

con = sqlite3.connect("student.db")
```

If `student.db` doesn't exist, SQLite will normally create it.

You can also create an in-memory database:

```python
con = sqlite3.connect(":memory:")
```

This database exists only while the program is running.

---

# 2. `connection.cursor()`

### Definition:

**`cursor()` creates a cursor object used to execute SQL commands.**

```python
cur = con.cursor()
```

Think of the cursor as the object through which Python sends SQL commands to the database.

---

# 3. `cursor.execute()`

### Definition:

**`execute()` executes an SQL statement.**

For example:

```python
cur.execute("""
    CREATE TABLE student (
        id INTEGER,
        name TEXT,
        age INTEGER
    )
""")
```

This creates a table called `student`.

---

# 🏗️ Creating a Table

Complete example:

```python
import sqlite3

con = sqlite3.connect("student.db")

cur = con.cursor()

cur.execute("""
CREATE TABLE student (
    id INTEGER,
    name TEXT,
    age INTEGER
)
""")

con.commit()

con.close()
```

The database will contain:

```text
student
----------------
id
name
age
```

---

# 4. `CREATE TABLE`

This is SQL rather than a Python function.

### Definition:

**`CREATE TABLE` creates a new table in the database.**

```python
cur.execute("""
CREATE TABLE employee (
    id INTEGER,
    name TEXT,
    salary REAL
)
""")
```

---

# 5. `IF NOT EXISTS`

Suppose you run your program twice.

The second time, the table already exists and you may get an error.

Use:

```python
cur.execute("""
CREATE TABLE IF NOT EXISTS student (
    id INTEGER,
    name TEXT,
    age INTEGER
)
""")
```

### Definition:

**`IF NOT EXISTS` prevents an error when the table already exists.**

This is very commonly used.

---

# ➕ Inserting Data

# 6. `INSERT INTO`

### Definition:

**`INSERT INTO` adds a new record to a table.**

```python
cur.execute("""
INSERT INTO student VALUES (1, 'Ayush', 25)
""")

con.commit()
```

Now the table contains:

```text
id    name     age
-------------------
1     Ayush     25
```

---

# 7. Parameterized Query

Instead of directly putting values into SQL, use `?`.

```python
name = "Rahul"
age = 24

cur.execute(
    "INSERT INTO student (name, age) VALUES (?, ?)",
    (name, age)
)

con.commit()
```

This is the **recommended approach** because it helps prevent SQL injection and handles values safely.

---

# 📖 Reading Data

# 8. `SELECT`

### Definition:

**`SELECT` retrieves data from a table.**

```python
cur.execute("SELECT * FROM student")
```

But this only executes the query. We need to retrieve the results.

---

# 9. `fetchone()`

### Definition:

**`fetchone()` retrieves one row from the query result.**

```python
cur.execute("SELECT * FROM student")

row = cur.fetchone()

print(row)
```

Output:

```text
(1, 'Ayush', 25)
```

---

# 10. `fetchall()`

### Definition:

**`fetchall()` retrieves all remaining rows from the query result.**

```python
cur.execute("SELECT * FROM student")

rows = cur.fetchall()

print(rows)
```

Output:

```text
[(1, 'Ayush', 25), (2, 'Rahul', 24)]
```

You can use a loop:

```python
for row in rows:
    print(row)
```

Output:

```text
(1, 'Ayush', 25)
(2, 'Rahul', 24)
```

---

# 11. `fetchmany()`

### Definition:

**`fetchmany(n)` retrieves the next `n` rows.**

```python
cur.execute("SELECT * FROM student")

rows = cur.fetchmany(2)

print(rows)
```

If there are many records, it retrieves only the first two.

---

# ✏️ Updating Data

# 12. `UPDATE`

### Definition:

**`UPDATE` modifies existing records in a table.**

Suppose:

```text
1 | Ayush | 25
```

We want to change the age:

```python
cur.execute(
    "UPDATE student SET age = ? WHERE id = ?",
    (26, 1)
)

con.commit()
```

Now:

```text
1 | Ayush | 26
```

---

# ❌ Deleting Data

# 13. `DELETE`

### Definition:

**`DELETE` removes records from a table.**

```python
cur.execute(
    "DELETE FROM student WHERE id = ?",
    (1,)
)

con.commit()
```

The student with ID `1` is deleted.

⚠️ Be careful:

```python
DELETE FROM student
```

without a `WHERE` condition can delete **all records**.

---

# 💾 `commit()`

### Definition:

**`commit()` permanently saves changes made to the database.**

```python
con.commit()
```

You normally call it after:

```text
INSERT
UPDATE
DELETE
```

Example:

```python
cur.execute("INSERT INTO student VALUES (1, 'Ayush', 25)")

con.commit()
```

---

# 🔒 `rollback()`

### Definition:

**`rollback()` cancels changes that have not yet been committed.**

```python
cur.execute(
    "UPDATE student SET age = 100 WHERE id = 1"
)

con.rollback()
```

The uncommitted update is cancelled.

---

# ❎ `close()`

### Definition:

**`close()` closes the database connection.**

```python
con.close()
```

Always close the connection when you're finished.

---

# 🔢 `lastrowid`

### Definition:

**`lastrowid` gives the ID of the most recently inserted row when applicable.**

For example:

```python
cur.execute(
    "INSERT INTO student (name, age) VALUES (?, ?)",
    ("Ayush", 25)
)

print(cur.lastrowid)

con.commit()
```

---

# 🔢 Using `AUTOINCREMENT`

You can let SQLite automatically generate IDs.

```python
cur.execute("""
CREATE TABLE IF NOT EXISTS student (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    age INTEGER
)
""")
```

Then:

```python
cur.execute(
    "INSERT INTO student (name, age) VALUES (?, ?)",
    ("Ayush", 25)
)

con.commit()
```

SQLite automatically generates:

```text
id = 1
```

Next student:

```text
id = 2
```

---

# 🔍 Using `WHERE`

### Definition:

**`WHERE` specifies which records should be affected or returned.**

```python
cur.execute(
    "SELECT * FROM student WHERE age > ?",
    (20,)
)
```

This returns students whose age is greater than 20.

---

# 📊 `ORDER BY`

### Definition:

**`ORDER BY` sorts the query results.**

```python
cur.execute(
    "SELECT * FROM student ORDER BY age"
)
```

For descending order:

```python
cur.execute(
    "SELECT * FROM student ORDER BY age DESC"
)
```

---

# 🔢 `COUNT()`

### Definition:

**`COUNT()` counts the number of records.**

```python
cur.execute("SELECT COUNT(*) FROM student")

result = cur.fetchone()

print(result)
```

Output:

```text
(5,)
```

Meaning there are 5 students.

---

# 🗑️ `DROP TABLE`

### Definition:

**`DROP TABLE` permanently removes a table and its data.**

```python
cur.execute("DROP TABLE student")
```

⚠️ Use carefully because the table is deleted.

---

# 🧠 Complete CRUD Example

CRUD means:

```text
C → Create
R → Read
U → Update
D → Delete
```

```python
import sqlite3

# Connection
con = sqlite3.connect("student.db")

# Cursor
cur = con.cursor()

# CREATE
cur.execute("""
CREATE TABLE IF NOT EXISTS student (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    age INTEGER
)
""")

# INSERT
cur.execute(
    "INSERT INTO student (name, age) VALUES (?, ?)",
    ("Ayush", 25)
)

# SAVE
con.commit()

# READ
cur.execute("SELECT * FROM student")

rows = cur.fetchall()

for row in rows:
    print(row)

# UPDATE
cur.execute(
    "UPDATE student SET age = ? WHERE name = ?",
    (26, "Ayush")
)

con.commit()

# DELETE
cur.execute(
    "DELETE FROM student WHERE name = ?",
    ("Ayush",)
)

con.commit()

# CLOSE
con.close()
```
