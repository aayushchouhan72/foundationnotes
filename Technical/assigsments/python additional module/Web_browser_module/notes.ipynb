# 🌐 `webbrowser` Module in Python

### What is `webbrowser`?

**`webbrowser` is a built-in Python module used to open web pages and URLs in your computer's default web browser.**

You don't need to install it.

```python
import webbrowser
```

For example, Python can open:

```text
Google
YouTube
GitHub
Any website
```

---

# 1. `webbrowser.open()`

### Definition:

**`open()` opens a URL in the default web browser.**

```python
import webbrowser

webbrowser.open("https://www.google.com")
```

Your default browser will open Google.

---

# 2. `webbrowser.open_new()`

### Definition:

**`open_new()` opens the URL in a new browser window if possible.**

```python
import webbrowser

webbrowser.open_new("https://www.google.com")
```

---

# 3. `webbrowser.open_new_tab()`

### Definition:

**`open_new_tab()` opens the URL in a new browser tab.**

```python
import webbrowser

webbrowser.open_new_tab("https://www.youtube.com")
```

This is useful when you want to open a website without replacing the current tab.

---

# 🔄 Difference Between `open()`, `open_new()` and `open_new_tab()`

| Function         | Purpose              |
| ---------------- | -------------------- |
| `open()`         | Open URL             |
| `open_new()`     | Open in a new window |
| `open_new_tab()` | Open in a new tab    |

Example:

```python
webbrowser.open("https://google.com")
```

```python
webbrowser.open_new("https://google.com")
```

```python
webbrowser.open_new_tab("https://google.com")
```

---

# 4. `webbrowser.get()`

### Definition:

**`get()` returns a browser controller that can be used to open URLs with a particular browser.**

```python
import webbrowser

browser = webbrowser.get()

browser.open("https://www.google.com")
```

The default browser is used.

---

# 5. Selecting a Specific Browser

You can request a particular browser using `get()` if that browser is registered/available on your system.

For example:

```python
import webbrowser

browser = webbrowser.get("chrome")
browser.open("https://www.google.com")
```

The exact browser name depends on your operating system and how the browser is registered.

---

# 6. `webbrowser.register()`

### Definition:

**`register()` registers a browser so that Python can use it through the `webbrowser` module.**

Basic form:

```python
webbrowser.register(
    "mybrowser",
    None,
    webbrowser.BackgroundBrowser("path-to-browser")
)
```

For beginners, you generally **don't need this function**.

---

# 🔥 Practical Example

Suppose you want a Python program that opens several websites:

```python
import webbrowser

webbrowser.open_new_tab("https://www.google.com")
webbrowser.open_new_tab("https://www.youtube.com")
webbrowser.open_new_tab("https://github.com")
```

Your browser will open these websites in tabs.

---

# 🎯 Mini Project: Website Launcher

You can create a simple menu:

```python
import webbrowser

print("1. Google")
print("2. YouTube")
print("3. GitHub")

choice = input("Enter your choice: ")

if choice == "1":
    webbrowser.open("https://www.google.com")

elif choice == "2":
    webbrowser.open("https://www.youtube.com")

elif choice == "3":
    webbrowser.open("https://github.com")

else:
    print("Invalid choice")
```

### Flow:

```text
             Start
               ↓
        Display websites
               ↓
        User selects 1/2/3
               ↓
       ┌───────┼────────┐
       ↓       ↓        ↓
    Google  YouTube   GitHub
       ↓       ↓        ↓
          Open Browser
```

# ⭐ Quick Revision

| Function                    | One-line meaning         |
| --------------------------- | ------------------------ |
| `webbrowser.open()`         | Open a URL               |
| `webbrowser.open_new()`     | Open URL in new window   |
| `webbrowser.open_new_tab()` | Open URL in new tab      |
| `webbrowser.get()`          | Get a browser controller |
| `webbrowser.register()`     | Register a browser       |

### 🧠 Easy way to remember

```text
webbrowser
    │
    ├── open()          → Open website
    ├── open_new()      → New window
    ├── open_new_tab()  → New tab
    │
    └── get()           → Get browser
```

**Important:** `webbrowser` only tells your installed browser to open a URL. It is **not** a web-scraping or HTTP-request library. For those tasks, you'd typically look at `urllib`, `requests`, or browser-automation tools.
