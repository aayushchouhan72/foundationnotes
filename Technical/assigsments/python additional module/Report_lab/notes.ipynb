# 📄 ReportLab Module in Python

### What is ReportLab?

**ReportLab is a Python library used to create PDF files, add text, tables, images, shapes, and other content to PDFs.**

Install it:

```bash
pip install reportlab
```

Import it:

```python
from reportlab.pdfgen import canvas
```

---

# 1. `canvas.Canvas()`

### Definition:

**`Canvas()` creates a PDF document that you can draw/write on.**

```python
from reportlab.pdfgen import canvas

pdf = canvas.Canvas("example.pdf")
```

This creates:

```text
example.pdf
```

---

# 2. `drawString()`

### Definition:

**`drawString()` writes text at a specified `(x, y)` position in the PDF.**

```python
from reportlab.pdfgen import canvas

pdf = canvas.Canvas("example.pdf")

pdf.drawString(100, 700, "Hello Python")

pdf.save()
```

The PDF will contain:

```text
Hello Python
```

### Understanding coordinates

PDF coordinates start from the **bottom-left corner**:

```text
(0, 842) ───────────────────
   │
   │
   │       Text
   │
   │
(0, 0) ─────────────────────
```

# 3. `save()`

### Definition:

**`save()` saves the PDF file.**

```python
pdf.save()
```

⚠️ Always call `save()` after finishing the PDF.

# 4. `setFont()`

### Definition:

**`setFont()` changes the font and font size used for text.**

```python
pdf.setFont("Helvetica", 20)

pdf.drawString(100, 700, "Hello Python")
```

```text
Helvetica
Times-Roman
Courier
```

# 5. `setFillColor()`

### Definition:

**`setFillColor()` changes the color of text or filled shapes.**

```python
from reportlab.lib.colors import red

pdf.setFillColor(red)

pdf.drawString(100, 700, "Red Text")
```

You can also use:

```python
from reportlab.lib import colors

pdf.setFillColor(colors.blue)
```

### Definition:

# 🔥 Complete Report Example

Here's a simple student report:

```python
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    Table
)
from reportlab.lib.styles import getSampleStyleSheet

doc = SimpleDocTemplate("student_report.pdf")

styles = getSampleStyleSheet()

elements = []

# Heading
heading = Paragraph(
    "Student Report",
    styles["Title"]
)

elements.append(heading)

# Space
elements.append(Spacer(1, 20))

# Table
data = [
    ["ID", "Name", "Age"],
    [1, "Ayush", 25],
    [2, "Rahul", 24],
    [3, "Amit", 23]
]

table = Table(data)

elements.append(table)

# Build PDF
doc.build(elements)
```

This creates:

```text
student_report.pdf

       Student Report

-------------------------
 ID    Name       Age
-------------------------
 1     Ayush      25
 2     Rahul      24
 3     Amit       23
-------------------------
```
