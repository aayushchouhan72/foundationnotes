# 🖼️ Pillow (PIL) in Python

### What is Pillow?

**Pillow is a Python library used to open, create, edit, resize, crop, rotate, and save images.**

Install it using:

```bash
pip install pillow
```

Import it as:

```python
from PIL import Image
```

> `PIL` means **Python Imaging Library**. Pillow is its modern, maintained version.

---

# 1. `Image.open()`

### Definition:

**`Image.open()` opens an existing image file.**

```python
from PIL import Image

img = Image.open("photo.jpg")

print(img)
```

Output may look like:

```text
<PIL.JpegImagePlugin.JpegImageFile image mode=RGB size=800x600>
```

# 2. `img.show()`

### Definition:

**`show()` opens the image using the computer's default image viewer.**

```python
from PIL import Image

img = Image.open("photo.jpg")

img.show()
```

# 9. `Image.new()`

### Definition:

**`Image.new()` creates a new blank image.**

```python
from PIL import Image

img = Image.new(
    "RGB",
    (500, 300),
    "white"
)

img.show()
```

Here:

```text
RGB       → color mode
(500,300) → width × height
white     → background color
```

---

# 10. `resize()`

### Definition:

**`resize()` changes the dimensions of an image.**

```python
from PIL import Image

img = Image.open("photo.jpg")

new_img = img.resize((400, 300))

new_img.show()
```

Original:

```text
800 × 600
```

New:

```text
400 × 300
```

---

# 11. `crop()`

### Definition:

**`crop()` cuts out a rectangular portion of an image.**

```python
img = Image.open("photo.jpg")

cropped = img.crop((100, 100, 400, 300))

cropped.show()
```

The coordinates are:

```text
(left, top, right, bottom)
```

---

# 12. `rotate()`

### Definition:

**`rotate()` rotates an image by a specified number of degrees.**

```python
img = Image.open("photo.jpg")

rotated = img.rotate(90)

rotated.show()
```

This rotates the image by **90 degrees**.

---

# 13. `transpose()`

### Definition:

**`transpose()` flips or rotates an image using predefined transformation methods.**

Flip horizontally:

```python
flipped = img.transpose(Image.Transpose.FLIP_LEFT_RIGHT)

flipped.show()
```

Flip vertically:

```python
flipped = img.transpose(Image.Transpose.FLIP_TOP_BOTTOM)

flipped.show()
```

---

# 14. `convert()`

### Definition:

**`convert()` changes an image from one color mode to another.**

Convert to grayscale:

```python
img = Image.open("photo.jpg")

gray = img.convert("L")

gray.show()
```

Here:

```text
RGB → L
```

`L` means grayscale.

---

# 15. `thumbnail()`

### Definition:

**`thumbnail()` reduces an image while maintaining its aspect ratio.**

```python
img = Image.open("photo.jpg")

img.thumbnail((300, 300))

img.show()
```

Unlike `resize()`, `thumbnail()` automatically maintains the image's proportions.

---

# 16. `ImageOps.grayscale()`

### Definition:

**`ImageOps.grayscale()` converts an image into grayscale.**

```python
from PIL import Image, ImageOps

img = Image.open("photo.jpg")

gray = ImageOps.grayscale(img)

gray.show()
```

---

# 17. `ImageOps.flip()`

### Definition:

**`ImageOps.flip()` flips an image vertically.**

```python
from PIL import Image, ImageOps

img = Image.open("photo.jpg")

result = ImageOps.flip(img)

result.show()
```

---

# 18. `ImageOps.mirror()`

### Definition:

**`ImageOps.mirror()` flips an image horizontally.**

```python
from PIL import Image, ImageOps

img = Image.open("photo.jpg")

result = ImageOps.mirror(img)

result.show()
```

# 🎨 Drawing on Images

Pillow also provides the `ImageDraw` module.

```python
from PIL import Image, ImageDraw
```

---

# 19. `ImageDraw.Draw()`

### Definition:

**`ImageDraw.Draw()` creates a drawing object that allows you to draw shapes and text on an image.**

```python
from PIL import Image, ImageDraw

img = Image.new("RGB", (500, 300), "white")

draw = ImageDraw.Draw(img)
```

Now `draw` can be used to draw objects.

---

# 20. `draw.rectangle()`

### Definition:

**`rectangle()` draws a rectangle on an image.**

```python
draw.rectangle(
    (100, 50, 300, 200),
    outline="red",
    width=5
)
```

---

# 21. `draw.ellipse()`

### Definition:

**`ellipse()` draws an ellipse or circle.**

```python
draw.ellipse(
    (100, 50, 300, 200),
    outline="blue",
    width=5
)
```

---

# 22. `draw.line()`

### Definition:

**`line()` draws a line between points.**

```python
draw.line(
    (50, 50, 400, 250),
    fill="black",
    width=5
)
```

---

# 23. `draw.text()`

### Definition:

**`text()` writes text onto an image.**

```python
draw.text(
    (100, 100),
    "Hello Python",
    fill="black"
)
```

Then:

```python
img.show()
```

# 🔍 Pixel Access

You can also work with individual pixels.

# 24. `getpixel()`

### Definition:

**`getpixel()` returns the color value of a specific pixel.**

```python
img = Image.open("photo.jpg")

pixel = img.getpixel((100, 100))

print(pixel)
```

For an RGB image, output could be:

```text
(120, 200, 50)
```

---

# 25. `putpixel()`

### Definition:

**`putpixel()` changes the color of a specific pixel.**

```python
img.putpixel((100, 100), (255, 0, 0))
```

The pixel at `(100,100)` becomes red.

---

# 🔥 Complete Example

Let's create an image, draw a rectangle and circle, and add text.

```python
from PIL import Image, ImageDraw

# Create image
img = Image.new("RGB", (600, 400), "white")

# Create drawing object
draw = ImageDraw.Draw(img)

# Rectangle
draw.rectangle(
    (100, 100, 300, 250),
    outline="red",
    width=5
)

# Circle
draw.ellipse(
    (350, 100, 500, 250),
    outline="blue",
    width=5
)

# Text
draw.text(
    (220, 300),
    "Python Image",
    fill="black"
)

# Show
img.show()

# Save
img.save("result.png")
```

This creates an image containing:

```text
┌───────────────────────────────┐
│                               │
│   ┌─────────┐       ○         │
│   │         │                 │
│   │         │                 │
│   └─────────┘                 │
│                               │
│       Python Image            │
└───────────────────────────────┘
```
