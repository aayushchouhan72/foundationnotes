# 🗣️ `pyttsx3` Module in Python

### What is `pyttsx3`?

**`pyttsx3` is a Python library used to convert text into speech (TTS — Text-to-Speech).**

For example:

```text
Python text
    ↓
pyttsx3
    ↓
🔊 Computer speaks the text
```

It can work **offline**, unlike services that require an internet connection.

---

# 1. Installation

`pyttsx3` is not built into Python.

Install it using:

```bash
pip install pyttsx3
```

Then import it:

```python
import pyttsx3
```

---

# 2. `pyttsx3.init()`

### Definition:

**`init()` initializes the text-to-speech engine.**

```python
import pyttsx3

engine = pyttsx3.init()
```

Now `engine` is the object we use to control speech.

---

# 3. `engine.say()`

### Definition:

**`say()` adds text that you want the computer to speak.**

```python
import pyttsx3

engine = pyttsx3.init()

engine.say("Hello, welcome to Python")
```

⚠️ `say()` normally **queues the speech**; it doesn't by itself start speaking immediately.

---

# 4. `engine.runAndWait()`

### Definition:

**`runAndWait()` processes the queued speech and makes the computer speak.**

```python
import pyttsx3

engine = pyttsx3.init()

engine.say("Hello Python")
engine.runAndWait()
```

🔊 Your computer will say:

> Hello Python

### Important:

```python
engine.say()
```

→ What should be spoken

```python
engine.runAndWait()
```

→ Actually process/speak it

---

# 5. Speaking Multiple Sentences

You can queue multiple messages:

```python
import pyttsx3

engine = pyttsx3.init()

engine.say("Hello")
engine.say("My name is Python")
engine.say("I can speak")

engine.runAndWait()
```

The engine processes them in order.

# 🎚️ Voice Properties

`pyttsx3` allows you to control how the speech sounds.

---

# 6. `engine.setProperty()`

### Definition:

**`setProperty()` changes a property of the speech engine, such as speed, volume, or voice.**

Basic structure:

```python
engine.setProperty("property", value)
```

For example:

```python
engine.setProperty("rate", 150)
```

---

# 7. `rate`

### Definition:

**`rate` controls the speaking speed.**

```python
import pyttsx3

engine = pyttsx3.init()

engine.setProperty("rate", 150)

engine.say("This is a slow speaking speed")
engine.runAndWait()
```

You can try:

```python
engine.setProperty("rate", 100)
```

or:

```python
engine.setProperty("rate", 200)
```

Higher value generally means faster speech.

---

# 8. `volume`

### Definition:

**`volume` controls the speech volume.**

The typical range is:

```text
0.0 → Silent
1.0 → Maximum
```

Example:

```python
engine.setProperty("volume", 1.0)

engine.say("Hello")
engine.runAndWait()
```

Another example:

```python
engine.setProperty("volume", 0.5)
```

This sets the volume to approximately 50%.

---

# 🎤 Voices

Your computer may have multiple installed speech voices.

# 9. `engine.getProperty("voices")`

### Definition:

**`getProperty("voices")` returns the voices available to the speech engine.**

```python
import pyttsx3

engine = pyttsx3.init()

voices = engine.getProperty("voices")

for voice in voices:
    print(voice.id)
```

You might see something like:

```text
HKEY_LOCAL_MACHINE\...
HKEY_LOCAL_MACHINE\...
```

The exact voices depend on your operating system and installed speech engines.

---

# 10. Changing Voice

You can select one of the available voices.

```python
voices = engine.getProperty("voices")

engine.setProperty("voice", voices[0].id)

engine.say("Hello Python")
engine.runAndWait()
```

Another voice:

```python
engine.setProperty("voice", voices[1].id)
```

⚠️ `voices[1]` only works if your system actually has a second voice.

---

# 11. `engine.getProperty("rate")`

### Definition:

**`getProperty("rate")` returns the current speech rate.**

```python
rate = engine.getProperty("rate")

print(rate)
```

---

# 12. `engine.getProperty("volume")`

### Definition:

**`getProperty("volume")` returns the current speech volume.**

```python
volume = engine.getProperty("volume")

print(volume)
```

---

# 13. `engine.stop()`

### Definition:

**`stop()` stops the speech engine and clears pending speech.**

```python
engine.say("This is a very long sentence...")

engine.stop()
```

This is useful when you want to interrupt speech.

---

# 🔥 Complete Example

Let's create a simple voice assistant:

```python
import pyttsx3

engine = pyttsx3.init()

# Set speed
engine.setProperty("rate", 150)

# Set volume
engine.setProperty("volume", 1.0)

# Speak
engine.say("Hello!")
engine.say("Welcome to Python.")
engine.say("I am your voice assistant.")

# Start speaking
engine.runAndWait()
```

# 🎯 Taking Input and Speaking It

You can combine `input()` with `pyttsx3`:

```python
import pyttsx3

engine = pyttsx3.init()

text = input("Enter something: ")

engine.say(text)
engine.runAndWait()
```

# 🔥 Mini Project: Talking Calculator

You can combine it with your Python basics:

```python
import pyttsx3

engine = pyttsx3.init()

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

result = a + b

print("Result:", result)

engine.say(f"The result is {result}")
engine.runAndWait()
```
