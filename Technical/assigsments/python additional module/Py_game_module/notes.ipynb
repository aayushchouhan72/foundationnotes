# 🎮 Pygame Module in Python

### What is Pygame?

**Pygame is a Python library used to create 2D games, animations, graphics, sounds, and interactive applications.**

Unlike `calendar`, `datetime`, or `urllib`, **Pygame is not built into Python**, so you normally install it first:

```bash
pip install pygame
```

Then:

```python
import pygame
```

# 🔄 The Most Important Concept: Game Loop

Almost every Pygame game has a **game loop**.

Basic structure:

```python
import pygame

pygame.init()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("My Game")

clock = pygame.time.Clock()

running = True

while running:

    # 1. Handle events
    for event in pygame.event.get():

        if event.type == pygame.QUIT:
            running = False

    # 2. Update game
    # player movement
    # enemy movement
    # score etc.

    # 3. Draw
    screen.fill((0, 0, 0))

    pygame.draw.circle(
        screen,
        (255, 0, 0),
        (400, 300),
        50
    )

    # 4. Update screen
    pygame.display.flip()

    # 5. Control FPS
    clock.tick(60)

pygame.quit()
```

### Understand the flow:

```text
        START
          ↓
   pygame.init()
          ↓
     Create Window
          ↓
      GAME LOOP
          ↓
   ┌──────────────┐
   │ Handle Events │
   └──────┬───────┘
          ↓
   ┌──────────────┐
   │ Update Game  │
   └──────┬───────┘
          ↓
   ┌──────────────┐
   │ Draw Objects │
   └──────┬───────┘
          ↓
   display.flip()
          ↓
      clock.tick()
          ↓
       Repeat
          ↓
     User closes
          ↓
    pygame.quit()
```
