### ⭐ For learning LangChain

I'd start with:

**1. Gemini** → good free option  
**2. Groq** → excellent for fast inference  
**3. Mistral** → useful for learning another provider  
**4. OpenAI** → learn the API, but don't assume the API itself is free

Google currently documents a free Gemini API tier with limited models and free input/output tokens. [Google AI for Developers](https://ai.google.dev/gemini-api/docs/pricing.md?utm_source=chatgpt.com) Groq provides an API-key console and project-based rate limits. [GroqCloud](https://console.groq.com/keys?utm_source=chatgpt.com)

For your `.env`, you can keep all four like this:

```env
GEMINI_API_KEY=your_gemini_key
OPENAI_API_KEY=your_openai_key
MISTRAL_API_KEY=your_mistral_key
GROQ_API_KEY=your_groq_key
```

**Don't share these keys in ChatGPT, GitHub, screenshots, or your code.** If a key is exposed, revoke/rotate it. Google specifically warns that leaked Gemini keys can be blocked. [Google AI for Developers](https://ai.google.dev/gemini-api/docs/troubleshooting?utm_source=chatgpt.com)
