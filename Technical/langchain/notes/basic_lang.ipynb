**4. What is LangChain?**

* LangChain is a framework for building applications powered by LLMs.
* It provides components for prompts, models, tools, retrieval, agents, etc.

**5. Basic Architecture**

```text
User
  ↓
Prompt
  ↓
LangChain
  ↓
Chat Model / LLM
  ↓
Response
```

**6. Direct LLM vs LangChain**

```text
Without LangChain

User → API → LLM → Response


With LangChain

User
 ↓
Prompt Template
 ↓
LangChain
 ↓
Chat Model
 ↓
Parser
 ↓
Response
```

**21. Advantages & Limitations**

Advantages:

* Modular
* Supports many model providers
* Prompt management
* RAG support
* Tools and agents
* Integrations

Limitations:

* Adds abstraction
* Framework APIs can change
* Debugging chains/agents can become complex
* LLM limitations still remain
