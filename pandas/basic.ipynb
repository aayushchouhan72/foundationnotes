```python
import pandas as pd

df = pd.DataFrame({
    "salary": [25000, 28000, 30000, 45000,32000, 35000, 31000, 29000, 500000]
})
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)

IQR = Q3 - Q1

print()

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df["salary"] >= lower) & (df["salary"] <= upper)]

df
```
